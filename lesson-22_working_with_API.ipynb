````python
import requests
import sqlite3
import smtplib
import schedule
import time
from datetime import datetime, timedelta
from email.mime.text import MIMEText  # Исправлено: MIMEText вместо MimeText
from email.mime.multipart import MIMEMultipart  # Исправлено: MIMEMultipart
from typing import Dict, List, Tuple
import json
from dataclasses import dataclass
from enum import Enum

# ============= ДОПОЛНИТЕЛЬНЫЕ БИБЛИОТЕКИ =============
# pip install requests schedule python-dotenv
# Для email: уже встроен smtplib

# ============= НАСТРОЙКИ =============
class UnitSystem(Enum):
    METRIC = "metric"      # °C, м/с
    IMPERIAL = "imperial"  # °F, мили/ч

class WindUnit(Enum):
    MS = "ms"    # метры в секунду
    KMH = "kmh"  # километры в час

@dataclass
class WeatherPreferences:
    temp_unit: UnitSystem = UnitSystem.METRIC
    wind_unit: WindUnit = WindUnit.MS
    favorite_cities: List[str] = None
    
    def __post_init__(self):
        if self.favorite_cities is None:
            self.favorite_cities = []

# ============= 1. ВЫБОР ЕДИНИЦ ИЗМЕРЕНИЯ =============
class UnitConverter:
    @staticmethod
    def convert_wind(speed_ms: float, unit: WindUnit) -> float:
        """Конвертация скорости ветра"""
        if unit == WindUnit.KMH:
            return speed_ms * 3.6  # м/с -> км/ч
        return speed_ms  # м/с
    
    @staticmethod
    def get_wind_unit_symbol(unit: WindUnit) -> str:
        return "км/ч" if unit == WindUnit.KMH else "м/с"
    
    @staticmethod
    def format_temperature(temp: float, system: UnitSystem) -> str:
        """Форматирование температуры"""
        if system == UnitSystem.IMPERIAL:
            # Конвертация в Фаренгейт: (°C × 9/5) + 32
            fahrenheit = (temp * 9/5) + 32
            return f"{fahrenheit:.1f}°F"
        return f"{temp:.1f}°C"

# ============= 2. ПРОГНОЗ НА 5 ДНЕЙ =============
class Forecast5Days:
    def __init__(self, api_key: str):
        self.api_key = api_key
        self.base_url = "https://api.openweathermap.org/data/2.5/forecast"
    
    def get_forecast(self, city: str, unit: UnitSystem = UnitSystem.METRIC) -> List[Dict]:
        """Получение прогноза на 5 дней (каждые 3 часа)"""
        params = {
            'q': city,
            'appid': self.api_key,
            'units': unit.value,
            'lang': 'ru'
        }
        
        try:
            response = requests.get(self.base_url, params=params, timeout=10)
            if response.status_code == 200:
                data = response.json()
                return self._process_forecast(data)
            else:
                print(f"❌ Ошибка получения прогноза: {response.status_code}")
                return []
        except Exception as e:
            print(f"❌ Ошибка: {e}")
            return []
    
    def _process_forecast(self, data: Dict) -> List[Dict]:
        """Обработка и группировка прогноза по дням"""
        forecast_by_day = {}
        
        for item in data['list']:
            date = item['dt_txt'].split()[0]  # YYYY-MM-DD
            if date not in forecast_by_day:
                forecast_by_day[date] = []
            forecast_by_day[date].append(item)
        
        # Агрегируем данные по дням
        daily_forecast = []
        for date, items in forecast_by_day.items():
            temps = [item['main']['temp'] for item in items]
            daily_forecast.append({
                'date': date,
                'temp_min': min(temps),
                'temp_max': max(temps),
                'description': items[0]['weather'][0]['description'],
                'humidity': items[0]['main']['humidity'],
                'wind_speed': items[0]['wind']['speed']
            })
        
        return daily_forecast[:5]  # Первые 5 дней
    
    def display_forecast(self, forecast: List[Dict], city: str, converter: UnitConverter, 
                        temp_unit: UnitSystem, wind_unit: WindUnit):
        """Красивый вывод прогноза"""
        print(f"\n{'='*60}")
        print(f"📅 ПРОГНОЗ НА 5 ДНЕЙ: {city.upper()}")
        print(f"{'='*60}")
        
        for day in forecast:
            date_obj = datetime.strptime(day['date'], '%Y-%m-%d')
            weekday = date_obj.strftime('%A')
            
            temp_min = converter.format_temperature(day['temp_min'], temp_unit)
            temp_max = converter.format_temperature(day['temp_max'], temp_unit)
            wind = converter.convert_wind(day['wind_speed'], wind_unit)
            wind_symbol = converter.get_wind_unit_symbol(wind_unit)
            
            print(f"\n📆 {weekday} ({day['date']})")
            print(f"   🌡️  {temp_min} / {temp_max}")
            print(f"   ☁️  {day['description'].capitalize()}")
            print(f"   💧 Влажность: {day['humidity']}%")
            print(f"   🌬️  Ветер: {wind:.1f} {wind_symbol}")
        
        print(f"{'='*60}\n")

# ============= 3. СОХРАНЕНИЕ ИСТОРИИ В SQLITE =============
class WeatherHistory:
    def __init__(self, db_path: str = "weather_history.db"):
        self.conn = sqlite3.connect(db_path)
        self.create_tables()
    
    def create_tables(self):
        """Создание таблиц для истории"""
        cursor = self.conn.cursor()
        
        # Таблица истории запросов
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS searches (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                city TEXT NOT NULL,
                timestamp DATETIME DEFAULT CURRENT_TIMESTAMP,
                temperature REAL,
                humidity INTEGER
            )
        ''')
        
        # Таблица для частых запросов (кэш)
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS popular_cities (
                city TEXT PRIMARY KEY,
                search_count INTEGER DEFAULT 1,
                last_searched DATETIME DEFAULT CURRENT_TIMESTAMP
            )
        ''')
        
        self.conn.commit()
    
    def save_search(self, city: str, temperature: float = None, humidity: int = None):
        """Сохранение поиска"""
        cursor = self.conn.cursor()
        
        # Сохраняем в историю
        cursor.execute('''
            INSERT INTO searches (city, temperature, humidity)
            VALUES (?, ?, ?)
        ''', (city, temperature, humidity))
        
        # Обновляем счетчик популярных городов
        cursor.execute('''
            INSERT INTO popular_cities (city, search_count)
            VALUES (?, 1)
            ON CONFLICT(city) DO UPDATE SET
                search_count = search_count + 1,
                last_searched = CURRENT_TIMESTAMP
        ''', (city,))
        
        self.conn.commit()
    
    def get_popular_cities(self, limit: int = 5) -> List[Tuple[str, int]]:
        """Получить самые частые запросы"""
        cursor = self.conn.cursor()
        cursor.execute('''
            SELECT city, search_count 
            FROM popular_cities 
            ORDER BY search_count DESC 
            LIMIT ?
        ''', (limit,))
        return cursor.fetchall()
    
    def get_search_stats(self) -> Dict:
        """Статистика поисков"""
        cursor = self.conn.cursor()
        
        # Общее количество запросов
        cursor.execute("SELECT COUNT(*) FROM searches")
        total = cursor.fetchone()[0]
        
        # Уникальные города
        cursor.execute("SELECT COUNT(DISTINCT city) FROM searches")
        unique = cursor.fetchone()[0]
        
        # Поиски за сегодня
        cursor.execute('''
            SELECT COUNT(*) FROM searches 
            WHERE DATE(timestamp) = DATE('now')
        ''')
        today = cursor.fetchone()[0]
        
        return {
            'total': total,
            'unique_cities': unique,
            'today': today
        }
    
    def show_history(self, limit: int = 10):
        """Показать последние поиски"""
        cursor = self.conn.cursor()
        cursor.execute('''
            SELECT city, timestamp, temperature, humidity 
            FROM searches 
            ORDER BY timestamp DESC 
            LIMIT ?
        ''', (limit,))
        
        results = cursor.fetchall()
        if results:
            print("\n📜 ПОСЛЕДНИЕ ЗАПРОСЫ:")
            print("-" * 50)
            for city, timestamp, temp, humidity in results:
                print(f"📍 {city} | {timestamp[:19]} | {temp}°C | 💧{humidity}%")
        else:
            print("\n📜 История поисков пуста")
    
    def close(self):
        self.conn.close()

# ============= 4. ГЕОЛОКАЦИЯ ПО IP =============
class GeoLocation:
    @staticmethod
    def get_city_by_ip() -> str:
        """Определение города по IP адресу"""
        try:
            # Бесплатный API для геолокации
            response = requests.get('http://ip-api.com/json/', timeout=5)
            if response.status_code == 200:
                data = response.json()
                if data['status'] == 'success':
                    city = data['city']
                    country = data['country']
                    print(f"📍 Определено ваше местоположение: {city}, {country}")
                    return city
            return None
        except Exception as e:
            print(f"❌ Не удалось определить местоположение: {e}")
            return None
    
    @staticmethod
    def get_weather_by_coords(lat: float, lon: float, api_key: str) -> Dict:
        """Получение погоды по координатам"""
        url = f"https://api.openweathermap.org/data/2.5/weather?lat={lat}&lon={lon}&appid={api_key}&units=metric&lang=ru"
        try:
            response = requests.get(url, timeout=10)
            if response.status_code == 200:
                return response.json()
        except:
            pass
        return None

# ============= 5. EMAIL-РАССЫЛКА =============
class EmailWeatherSender:
    def __init__(self, smtp_server: str = "smtp.gmail.com", smtp_port: int = 587):
        self.smtp_server = smtp_server
        self.smtp_port = smtp_port
        self.email_config = self.load_email_config()
    
    def load_email_config(self) -> Dict:
        """Загрузка конфигурации email из файла"""
        try:
            with open('email_config.json', 'r') as f:
                return json.load(f)
        except FileNotFoundError:
            return {
                'enabled': False,
                'sender_email': '',
                'sender_password': '',
                'recipient_email': ''
            }
    
    def save_email_config(self, sender_email: str, sender_password: str, recipient_email: str):
        """Сохранение настроек email"""
        self.email_config = {
            'enabled': True,
            'sender_email': sender_email,
            'sender_password': sender_password,
            'recipient_email': recipient_email
        }
        with open('email_config.json', 'w') as f:
            json.dump(self.email_config, f)
        print("✅ Настройки email сохранены!")
    
    def send_weather_email(self, city: str, weather_data: Dict, forecast_data: List[Dict]):
        """Отправка прогноза погоды на email"""
        if not self.email_config['enabled']:
            print("❌ Email рассылка не настроена")
            return False
        
        try:
            # Создаем письмо - ИСПРАВЛЕНО: MIMEMultipart
            msg = MIMEMultipart()
            msg['From'] = self.email_config['sender_email']
            msg['To'] = self.email_config['recipient_email']
            msg['Subject'] = f"🌤️ Прогноз погоды для {city} на сегодня"
            
            # Формируем HTML содержимое
            html_content = self._create_email_html(city, weather_data, forecast_data)
            # ИСПРАВЛЕНО: MIMEText
            msg.attach(MIMEText(html_content, 'html'))
            
            # Отправляем
            server = smtplib.SMTP(self.smtp_server, self.smtp_port)
            server.starttls()
            server.login(self.email_config['sender_email'], self.email_config['sender_password'])
            server.send_message(msg)
            server.quit()
            
            print(f"✅ Прогноз для {city} отправлен на {self.email_config['recipient_email']}")
            return True
            
        except Exception as e:
            print(f"❌ Ошибка отправки email: {e}")
            return False
    
    def _create_email_html(self, city: str, weather: Dict, forecast: List[Dict]) -> str:
        """Создание HTML письма с погодой"""
        html = f"""
        <html>
        <head>
            <style>
                body {{ font-family: Arial, sans-serif; }}
                .weather-box {{ background-color: #f0f0f0; padding: 20px; border-radius: 10px; }}
                .forecast-day {{ margin: 10px 0; padding: 10px; border-bottom: 1px solid #ddd; }}
            </style>
        </head>
        <body>
            <h2>🌍 Прогноз погоды для {city}</h2>
            <div class="weather-box">
                <h3>Текущая погода:</h3>
                <p>🌡️ Температура: {weather['main']['temp']:.1f}°C</p>
                <p>💧 Влажность: {weather['main']['humidity']}%</p>
                <p>🌬️ Ветер: {weather['wind']['speed']} м/с</p>
                <p>☁️ {weather['weather'][0]['description']}</p>
            </div>
            <h3>📅 Прогноз на 5 дней:</h3>
        """
        
        for day in forecast[:3]:  # Показываем 3 дня в email
            html += f"""
            <div class="forecast-day">
                <strong>{day['date']}</strong><br>
                🌡️ {day['temp_min']:.1f}°C - {day['temp_max']:.1f}°C<br>
                ☁️ {day['description']}<br>
                💧 Влажность: {day['humidity']}%
            </div>
            """
        
        html += """
        </body>
        </html>
        """
        return html

# ============= ГЛАВНОЕ ПРИЛОЖЕНИЕ =============
class WeatherAppPro:
    def __init__(self, api_key: str):
        self.api_key = api_key
        self.preferences = WeatherPreferences()
        self.converter = UnitConverter()
        self.history = WeatherHistory()
        self.email_sender = EmailWeatherSender()
        self.forecast_api = Forecast5Days(api_key)
        
        # Загрузка сохраненных предпочтений
        self.load_preferences()
    
    def load_preferences(self):
        """Загрузка предпочтений пользователя"""
        try:
            with open('weather_prefs.json', 'r') as f:
                prefs = json.load(f)
                self.preferences.temp_unit = UnitSystem(prefs.get('temp_unit', 'metric'))
                self.preferences.wind_unit = WindUnit(prefs.get('wind_unit', 'ms'))
                self.preferences.favorite_cities = prefs.get('favorite_cities', [])
            print("✅ Предпочтения загружены")
        except FileNotFoundError:
            print("📝 Используются настройки по умолчанию")
    
    def save_preferences(self):
        """Сохранение предпочтений"""
        prefs = {
            'temp_unit': self.preferences.temp_unit.value,
            'wind_unit': self.preferences.wind_unit.value,
            'favorite_cities': self.preferences.favorite_cities
        }
        with open('weather_prefs.json', 'w') as f:
            json.dump(prefs, f)
        print("✅ Настройки сохранены!")
    
    def configure_units(self):
        """Настройка единиц измерения"""
        print("\n⚙️ НАСТРОЙКА ЕДИНИЦ ИЗМЕРЕНИЯ:")
        print("1. Цельсий (°C) / Фаренгейт (°F)")
        temp_choice = input("Выберите (1-°C, 2-°F): ").strip()
        if temp_choice == '2':
            self.preferences.temp_unit = UnitSystem.IMPERIAL
        else:
            self.preferences.temp_unit = UnitSystem.METRIC
        
        print("\n2. Единицы измерения ветра:")
        print("1. м/с (метры в секунду)")
        print("2. км/ч (километры в час)")
        wind_choice = input("Выберите (1-м/с, 2-км/ч): ").strip()
        if wind_choice == '2':
            self.preferences.wind_unit = WindUnit.KMH
        else:
            self.preferences.wind_unit = WindUnit.MS
        
        self.save_preferences()
        print("✅ Единицы измерения обновлены!")
    
    def get_current_weather(self, city: str) -> Dict:
        """Получение текущей погоды"""
        url = "https://api.openweathermap.org/data/2.5/weather"
        params = {
            'q': city,
            'appid': self.api_key,
            'units': self.preferences.temp_unit.value,
            'lang': 'ru'
        }
        
        try:
            response = requests.get(url, params=params, timeout=10)
            if response.status_code == 200:
                data = response.json()
                # Сохраняем в историю
                self.history.save_search(
                    city, 
                    data['main']['temp'], 
                    data['main']['humidity']
                )
                return data
            else:
                print(f"❌ Город '{city}' не найден")
                return None
        except Exception as e:
            print(f"❌ Ошибка: {e}")
            return None
    
    def display_weather(self, data: Dict):
        """Отображение погоды с учетом единиц измерения"""
        city = data['name']
        temp = data['main']['temp']
        feels_like = data['main']['feels_like']
        humidity = data['main']['humidity']
        wind_speed = data['wind']['speed']
        description = data['weather'][0]['description']
        
        # Конвертируем скорость ветра если нужно
        wind_display = self.converter.convert_wind(wind_speed, self.preferences.wind_unit)
        wind_symbol = self.converter.get_wind_unit_symbol(self.preferences.wind_unit)
        
        print("\n" + "="*50)
        print(f"🌍 ПОГОДА: {city.upper()}")
        print("="*50)
        print(f"🌡️  Температура: {self.converter.format_temperature(temp, self.preferences.temp_unit)}")
        print(f"🤔 Ощущается как: {self.converter.format_temperature(feels_like, self.preferences.temp_unit)}")
        print(f"💧 Влажность: {humidity}%")
        print(f"🌬️  Ветер: {wind_display:.1f} {wind_symbol}")
        print(f"☁️  {description.capitalize()}")
        print("="*50)
    
    def show_popular_cities(self):
        """Показать популярные города"""
        popular = self.history.get_popular_cities()
        if popular:
            print("\n⭐ ПОПУЛЯРНЫЕ ГОРОДА:")
            for city, count in popular:
                print(f"   📍 {city}: {count} запросов")
        else:
            print("\n⭐ Нет данных о популярных городах")
    
    def show_stats(self):
        """Показать статистику"""
        stats = self.history.get_search_stats()
        print("\n📊 СТАТИСТИКА ИСПОЛЬЗОВАНИЯ:")
        print(f"   📝 Всего запросов: {stats['total']}")
        print(f"   🏙️  Уникальных городов: {stats['unique_cities']}")
        print(f"   📅 Сегодня: {stats['today']}")
    
    def setup_email_reminder(self):
        """Настройка email рассылки"""
        print("\n📧 НАСТРОЙКА EMAIL РАССЫЛКИ")
        print("(Для Gmail нужно включить 'Доступ для ненадежных приложений')")
        
        sender = input("Ваш email (отправитель): ").strip()
        password = input("Пароль приложения: ").strip()
        recipient = input("Email для получения прогнозов: ").strip()
        
        self.email_sender.save_email_config(sender, password, recipient)
        
        # Добавляем город для рассылки
        city = input("\nВведите город для ежедневной рассылки: ").strip()
        if city not in self.preferences.favorite_cities:
            self.preferences.favorite_cities.append(city)
            self.save_preferences()
        
        print("✅ Рассылка настроена! Прогноз будет приходить ежедневно в 8:00")
    
    def send_daily_forecast(self):
        """Отправка ежедневного прогноза (для schedule)"""
        for city in self.preferences.favorite_cities:
            weather = self.get_current_weather(city)
            if weather:
                forecast = self.forecast_api.get_forecast(city, self.preferences.temp_unit)
                if forecast:
                    self.email_sender.send_weather_email(city, weather, forecast)
    
    def start_scheduler(self):
        """Запуск планировщика для email рассылки"""
        if self.preferences.favorite_cities:
            schedule.every().day.at("08:00").do(self.send_daily_forecast)
            print("⏰ Запланирована ежедневная рассылка в 8:00")
            
            # Запускаем в отдельном потоке (для демонстрации)
            import threading
            def run_scheduler():
                while True:
                    schedule.run_pending()
                    time.sleep(60)
            
            thread = threading.Thread(target=run_scheduler, daemon=True)
            thread.start()
    
    def menu(self):
        """Главное меню"""
        print("\n" + "="*50)
        print("🌤️  ПОГОДНОЕ ПРИЛОЖЕНИЕ PRO")
        print("="*50)
        print("1. Погода сейчас")
        print("2. Прогноз на 5 дней")
        print("3. Настройки единиц измерения")
        print("4. История поисков")
        print("5. Популярные города")
        print("6. Статистика")
        print("7. Погода по IP (геолокация)")
        print("8. Настройка email рассылки")
        print("9. Показать избранные города")
        print("0. Выход")
        print("="*50)
        
        return input("Выберите действие: ").strip()
    
    def run(self):
        """Запуск приложения"""
        print("\n🌤️  Добро пожаловать в WeatherApp Pro!")
        
        # Запускаем планировщик
        self.start_scheduler()
        
        while True:
            choice = self.menu()
            
            if choice == '0':
                self.history.close()
                print("👋 До свидания!")
                break
            
            elif choice == '1':  # Погода сейчас
                city = input("Введите город: ").strip()
                weather_data = self.get_current_weather(city)
                if weather_data:
                    self.display_weather(weather_data)
            
            elif choice == '2':  # Прогноз на 5 дней
                city = input("Введите город: ").strip()
                forecast = self.forecast_api.get_forecast(city, self.preferences.temp_unit)
                if forecast:
                    self.forecast_api.display_forecast(
                        forecast, city, self.converter,
                        self.preferences.temp_unit, self.preferences.wind_unit
                    )
            
            elif choice == '3':  # Настройки
                self.configure_units()
            
            elif choice == '4':  # История
                self.history.show_history()
            
            elif choice == '5':  # Популярные города
                self.show_popular_cities()
            
            elif choice == '6':  # Статистика
                self.show_stats()
            
            elif choice == '7':  # Геолокация
                city = GeoLocation.get_city_by_ip()
                if city:
                    weather_data = self.get_current_weather(city)
                    if weather_data:
                        self.display_weather(weather_data)
                else:
                    print("Не удалось определить местоположение")
            
            elif choice == '8':  # Email рассылка
                self.setup_email_reminder()
            
            elif choice == '9':  # Избранные города
                if self.preferences.favorite_cities:
                    print("\n⭐ ИЗБРАННЫЕ ГОРОДА:")
                    for city in self.preferences.favorite_cities:
                        print(f"   📍 {city}")
                else:
                    print("\n⭐ Нет избранных городов")

# ============= ЗАПУСК =============
if __name__ == "__main__":
    API_KEY = input("🔑 Введите API ключ OpenWeatherMap: ").strip()
    
    app = WeatherAppPro(API_KEY)
    app.run()
````

````python
import requests
import time

class WeatherApp:
    def __init__(self):
        print("🌤️  Настройка погодного приложения")
        print("="*50)
        print("Как получить API ключ OpenWeatherMap:")
        print("1. Перейдите на https://home.openweathermap.org/users/sign_up")
        print("2. Зарегистрируйтесь и подтвердите email")
        print("3. Перейдите в раздел 'API Keys'")
        print("4. Скопируйте ключ (он будет активирован через 10-30 минут)")
        print("="*50)
        
        self.api_key = input("\n🔑 Введите ваш API ключ: ").strip()
        self.base_url = "https://api.openweathermap.org/data/2.5/weather"
        
        # Проверка ключа
        self.test_api_key()
    
    def test_api_key(self):
        """Проверка валидности API ключа"""
        print("\n🔍 Проверяем API ключ...")
        
        test_params = {
            'q': 'London',
            'appid': self.api_key,
            'units': 'metric'
        }
        
        try:
            response = requests.get(self.base_url, params=test_params, timeout=10)
            
            if response.status_code == 200:
                print("✅ API ключ работает корректно!")
                return True
            elif response.status_code == 401:
                print("❌ Неверный API ключ!")
                print("\nВозможные причины:")
                print("- Ключ только что создан (подождите 10-30 минут)")
                print("- Вы не подтвердили email")
                print("- Ключ скопирован не полностью (лишние пробелы)")
                print("- Используете старый ключ")
                return False
            else:
                print(f"⚠️ Неожиданный ответ: {response.status_code}")
                return False
                
        except Exception as e:
            print(f"❌ Ошибка подключения: {e}")
            return False
    
    def get_weather(self, city_name):
        """Получение погоды"""
        params = {
            'q': city_name,
            'appid': self.api_key,
            'units': 'metric',
            'lang': 'ru'
        }
        
        try:
            response = requests.get(self.base_url, params=params, timeout=10)
            
            if response.status_code == 200:
                data = response.json()
                self.display_weather(data, city_name)
                return True
                
            elif response.status_code == 401:
                print("❌ Ошибка авторизации: API ключ недействителен")
                print("💡 Решение: получите новый ключ на сайте OpenWeatherMap")
                return False
                
            elif response.status_code == 404:
                print(f"❌ Город '{city_name}' не найден")
                print("💡 Совет: попробуйте ввести на английском")
                return False
                
            else:
                print(f"❌ Ошибка: {response.status_code}")
                return False
                
        except requests.exceptions.RequestException as e:
            print(f"❌ Ошибка сети: {e}")
            return False
    
    def display_weather(self, data, city_name):
        print("\n" + "="*50)
        print(f"📍 Погода в городе: {city_name.upper()}")
        print("="*50)
        print(f"🌡️  Температура: {data['main']['temp']:.1f}°C")
        print(f"💧 Влажность: {data['main']['humidity']}%")
        print(f"🌬️  Ветер: {data['wind']['speed']} м/с")
        print(f"☁️  {data['weather'][0]['description'].capitalize()}")
        print("="*50)
    
    def run(self):
        """Основной цикл"""
        print("\n" + "="*50)
        print("🌤️  Погодное приложение запущено!")
        print("Для выхода введите 'exit' или 'quit'")
        print("="*50)
        
        while True:
            city = input("\n🌆 Введите название города: ").strip()
            
            if city.lower() in ['exit', 'quit', 'выход', 'q']:
                print("👋 До свидания!")
                break
            
            if not city:
                print("⚠️ Пожалуйста, введите название города")
                continue
            
            success = self.get_weather(city)
            
            if not success:
                print("\n💡 Попробуйте:")
                print("- Проверить написание города (на английском)")
                print("- Ввести другой город")
                print("- Проверить API ключ командой 'key'")


# Запуск
if __name__ == "__main__":
    app = WeatherApp()
    
    # Если ключ не прошел проверку, предложим исправить
    if app.test_api_key():
        app.run()
    else:
        print("\n🔄 Хотите ввести другой API ключ? (да/нет)")
        if input().lower() in ['да', 'yes', 'y']:
            new_key = input("Введите новый ключ: ").strip()
            app.api_key = new_key
            if app.test_api_key():
                app.run()
````

````python
# 📚 ШАГ 5: Добавляем историю и избранное (упрощенно)
import requests
import json

class SimpleJokeApp:
    def __init__(self):
        self.history = []    # История просмотренных шуток
        self.favorites = []  # Избранные шутки
    
    def get_joke(self, category="Any", safe=True):
        """Получить шутку"""
        url = f"https://v2.jokeapi.dev/joke/{category}"
        params = {'blacklistFlags': 'nsfw'} if safe else {}
        
        response = requests.get(url, params=params)
        data = response.json()
        
        # Сохраняем в историю
        self.history.append(data)
        
        # Форматируем вывод
        if data['type'] == 'single':
            joke_text = data['joke']
        else:
            joke_text = f"{data['setup']}\n\n{data['delivery']}"
        
        return joke_text, data['id']
    
    def save_to_favorites(self, joke_id, joke_text, category):
        """Сохранить шутку в избранное"""
        self.favorites.append({
            'id': joke_id,
            'text': joke_text,
            'category': category
        })
        # Сохраняем в файл
        with open('favorites.json', 'w', encoding='utf-8') as f:
            json.dump(self.favorites, f, ensure_ascii=False, indent=2)
        print("✅ Шутка сохранена в избранное!")
    
    def show_favorites(self):
        """Показать избранное"""
        if not self.favorites:
            print("Избранное пусто")
            return
        
        print("\n⭐ ИЗБРАННЫЕ ШУТКИ:")
        for i, joke in enumerate(self.favorites, 1):
            print(f"\n{i}. [{joke['category']}]")
            print(f"   {joke['text'][:100]}...")
    
    def load_favorites(self):
        """Загрузить избранное из файла"""
        try:
            with open('favorites.json', 'r', encoding='utf-8') as f:
                self.favorites = json.load(f)
            print(f"Загружено {len(self.favorites)} избранных шуток")
        except:
            pass

# Использование
app = SimpleJokeApp()
app.load_favorites()

while True:
    print("\n" + "="*40)
    print("1. Случайная шутка")
    print("2. Избранное")
    print("3. Выход")
    
    choice = input("Выбор: ")
    
    if choice == "1":
        joke_text, joke_id = app.get_joke()
        print("\n" + joke_text)
        
        save = input("\nСохранить в избранное? (y/n): ")
        if save.lower() == 'y':
            app.save_to_favorites(joke_id, joke_text, "Random")
    
    elif choice == "2":
        app.show_favorites()
    
    elif choice == "3":
        break
````

````python
import requests
import random
import time
from typing import Dict, List, Optional, Tuple
from enum import Enum
import json

# ============= КАТЕГОРИИ ШУТОК =============
class JokeCategory(Enum):
    """Доступные категории шуток"""
    ANY = "Any"
    PROGRAMMING = "Programming"
    MISC = "Misc"
    DARK = "Dark"
    PUN = "Pun"
    SPOOKY = "Spooky"
    CHRISTMAS = "Christmas"
    
    @classmethod
    def get_all_categories(cls) -> List[str]:
        """Получить список всех категорий"""
        return [category.value for category in cls]
    
    @classmethod
    def get_categories_without_any(cls) -> List[str]:
        """Получить все категории кроме 'Any'"""
        return [category.value for category in cls if category != cls.ANY]

# ============= ТИПЫ ШУТОК =============
class JokeType(Enum):
    """Тип шутки (однострочная или с паузой)"""
    SINGLE = "single"
    TWOPART = "twopart"
    ANY = "any"

# ============= ОСНОВНОЙ КЛАСС ГЕНЕРАТОРА =============
class JokeGenerator:
    def __init__(self):
        self.base_url = "https://v2.jokeapi.dev/joke"
        self.history = []
        self.favorites = []
        self.blacklist_flags = ["nsfw", "religious", "political", "racist", "sexist", "explicit"]
        self.load_data()
    
    def load_data(self):
        """Загрузка сохраненных данных"""
        try:
            with open('joke_favorites.json', 'r', encoding='utf-8') as f:
                data = json.load(f)
                self.favorites = data.get('favorites', [])
                self.history = data.get('history', [])
            print("✅ Избранные шутки загружены")
        except FileNotFoundError:
            print("📝 Создана новая коллекция шуток")
    
    def save_data(self):
        """Сохранение избранных шуток"""
        data = {
            'favorites': self.favorites[-50:],  # Храним последние 50
            'history': self.history[-100:]      # Храним последние 100
        }
        with open('joke_favorites.json', 'w', encoding='utf-8') as f:
            json.dump(data, f, ensure_ascii=False, indent=2)
    
    def get_joke(self, category: JokeCategory = JokeCategory.ANY, 
                 joke_type: JokeType = JokeType.ANY,
                 blacklist: List[str] = None,
                 safe_mode: bool = True) -> Optional[Dict]:
        """
        Получение шутки из API
        
        Args:
            category: Категория шутки
            joke_type: Тип шутки (single/twopart)
            blacklist: Список флагов для исключения
            safe_mode: Режим безопасности (исключает NSFW)
        """
        # Формируем параметры запроса
        category_str = category.value
        
        # Базовые параметры
        params = {
            'format': 'json',
            'type': joke_type.value if joke_type != JokeType.ANY else 'twopart,single',
        }
        
        # Добавляем черный список
        if safe_mode or blacklist:
            flags = blacklist if blacklist else ['nsfw']
            if safe_mode:
                flags = ['nsfw', 'explicit']
            params['blacklistFlags'] = ','.join(flags)
        
        # Формируем URL с категорией
        url = f"{self.base_url}/{category_str}"
        
        try:
            response = requests.get(url, params=params, timeout=10)
            
            if response.status_code == 200:
                data = response.json()
                
                if data.get('error') == False:
                    # Сохраняем в историю
                    self.add_to_history(data)
                    return data
                else:
                    print(f"❌ Ошибка API: {data.get('message', 'Неизвестная ошибка')}")
                    return None
            else:
                print(f"❌ Ошибка HTTP: {response.status_code}")
                return None
                
        except requests.exceptions.RequestException as e:
            print(f"❌ Ошибка подключения: {e}")
            return None
    
    def add_to_history(self, joke: Dict):
        """Добавление шутки в историю"""
        joke_id = joke.get('id')
        if joke_id not in self.history:
            self.history.append(joke_id)
            self.save_data()
    
    def display_joke(self, joke: Dict):
        """Красивый вывод шутки"""
        if not joke:
            return
        
        print("\n" + "="*60)
        
        # Выводим категорию
        category = joke.get('category', 'Unknown')
        print(f"🎭 Категория: {category}")
        
        # Выводим флаги (если есть)
        flags = joke.get('flags', {})
        if flags:
            active_flags = [flag for flag, value in flags.items() if value]
            if active_flags:
                print(f"🏷️  Теги: {', '.join(active_flags)}")
        
        print("-" * 60)
        
        # Выводим шутку в зависимости от типа
        if joke.get('type') == 'single':
            print(f"📖 {joke['joke']}")
        else:
            print(f"😂 {joke['setup']}")
            time.sleep(1)  # Пауза для эффекта
            print(f"\n✨ {joke['delivery']}")
        
        print("-" * 60)
        
        # Дополнительная информация
        joke_id = joke.get('id')
        print(f"🔢 ID: {joke_id}")
        
        # Рейтинг (если есть)
        if joke.get('rating'):
            print(f"⭐ Рейтинг: {joke['rating']}/10")
        
        print("="*60)
    
    def get_random_joke_by_keyword(self, keyword: str) -> Optional[Dict]:
        """Поиск шутки по ключевому слову"""
        params = {
            'contains': keyword,
            'format': 'json'
        }
        
        url = f"{self.base_url}/Any"
        
        try:
            response = requests.get(url, params=params, timeout=10)
            if response.status_code == 200:
                data = response.json()
                if data.get('error') == False and data.get('joke'):
                    return data
            return None
        except:
            return None
    
    def add_to_favorites(self, joke: Dict):
        """Добавление в избранное"""
        joke_id = joke.get('id')
        
        # Проверяем, не сохранена ли уже шутка
        if any(fav['id'] == joke_id for fav in self.favorites):
            print("⚠️ Эта шутка уже в избранном!")
            return
        
        # Сохраняем шутку
        favorite_joke = {
            'id': joke_id,
            'category': joke.get('category'),
            'joke': joke.get('joke'),  # Для single
            'setup': joke.get('setup'),  # Для twopart
            'delivery': joke.get('delivery'),
            'type': joke.get('type'),
            'date_added': time.strftime("%Y-%m-%d %H:%M:%S")
        }
        
        self.favorites.append(favorite_joke)
        self.save_data()
        print("⭐ Шутка добавлена в избранное!")
    
    def show_favorites(self):
        """Показать избранные шутки"""
        if not self.favorites:
            print("\n📭 В избранном пока ничего нет")
            return
        
        print("\n" + "⭐" * 30)
        print("ИЗБРАННЫЕ ШУТКИ")
        print("⭐" * 30)
        
        for i, joke in enumerate(self.favorites, 1):
            print(f"\n{i}. [ID: {joke['id']}] Категория: {joke['category']}")
            if joke['type'] == 'single':
                print(f"   {joke['joke'][:100]}...")
            else:
                print(f"   {joke['setup'][:100]}...")
            print(f"   📅 Добавлено: {joke['date_added']}")
    
    def get_joke_by_id(self, joke_id: int) -> Optional[Dict]:
        """Получить шутку по ID"""
        url = f"{self.base_url}/Any"
        params = {'idRange': joke_id}
        
        try:
            response = requests.get(url, params=params, timeout=10)
            if response.status_code == 200:
                data = response.json()
                if data.get('error') == False:
                    return data
            return None
        except:
            return None
    
    def get_statistics(self):
        """Статистика использования"""
        print("\n📊 СТАТИСТИКА ГЕНЕРАТОРА ШУТОК:")
        print(f"   📝 Просмотрено шуток: {len(self.history)}")
        print(f"   ⭐ В избранном: {len(self.favorites)}")
        
        # Статистика по категориям
        if self.history:
            # Получаем категории из избранного
            categories = {}
            for joke in self.favorites:
                cat = joke['category']
                categories[cat] = categories.get(cat, 0) + 1
            
            if categories:
                print("\n   📊 Популярные категории в избранном:")
                for cat, count in sorted(categories.items(), key=lambda x: x[1], reverse=True):
                    print(f"      {cat}: {count} шт.")

# ============= КОНСОЛЬНЫЙ ИНТЕРФЕЙС =============
class JokeConsoleApp:
    def __init__(self):
        self.joker = JokeGenerator()
        self.current_category = JokeCategory.ANY
        self.safe_mode = True
    
    def show_menu(self):
        """Показать главное меню"""
        print("\n" + "🎭" * 30)
        print("ГЕНЕРАТОР ШУТОК v2.0")
        print("🎭" * 30)
        print(f"\n🛡️  Режим: {'БЕЗОПАСНЫЙ (без NSFW)' if self.safe_mode else 'ВСЕ ШУТКИ'}")
        print(f"📂 Текущая категория: {self.current_category.value}")
        print("\n📋 МЕНЮ:")
        print("1. 🎲 Случайная шутка")
        print("2. 📂 Выбрать категорию")
        print("3. 🔍 Шутка по ключевому слову")
        print("4. ⭐ Избранное")
        print("5. 🔢 Шутка по ID")
        print("6. 🎯 Несколько шуток подряд")
        print("7. 🛡️ Переключить безопасный режим")
        print("8. 📊 Статистика")
        print("9. ❤️ Рандом из избранного")
        print("0. 🚪 Выход")
        print("-" * 40)
    
    def select_category(self):
        """Выбор категории"""
        print("\n📂 ДОСТУПНЫЕ КАТЕГОРИИ:")
        categories = list(JokeCategory)
        
        for i, category in enumerate(categories, 1):
            print(f"{i}. {category.value}")
        
        print(f"{len(categories) + 1}. Случайная категория")
        
        choice = input("\nВыберите категорию: ").strip()
        
        try:
            choice_num = int(choice)
            if 1 <= choice_num <= len(categories):
                self.current_category = categories[choice_num - 1]
            elif choice_num == len(categories) + 1:
                self.current_category = random.choice(categories)
            else:
                print("❌ Неверный выбор")
                return
        except ValueError:
            print("❌ Введите число")
            return
        
        print(f"✅ Выбрана категория: {self.current_category.value}")
    
    def get_multiple_jokes(self):
        """Получить несколько шуток подряд"""
        try:
            count = int(input("\nСколько шуток показать (1-10): "))
            count = min(max(count, 1), 10)
        except ValueError:
            print("❌ Введите число")
            return
        
        print(f"\n🎲 Показываю {count} шуток...")
        
        for i in range(count):
            print(f"\n{'='*40}")
            print(f"ШУТКА {i+1}/{count}")
            print(f"{'='*40}")
            
            joke = self.joker.get_joke(
                self.current_category,
                safe_mode=self.safe_mode
            )
            
            if joke:
                self.joker.display_joke(joke)
                
                # Спрашиваем, добавить ли в избранное после каждой шутки
                if i < count - 1:  # Не спрашиваем после последней
                    add = input("\n⭐ Добавить эту шутку в избранное? (y/n): ").lower()
                    if add == 'y':
                        self.joker.add_to_favorites(joke)
                
                if i < count - 1:
                    input("\nНажмите Enter для следующей шутки...")
    
    def random_from_favorites(self):
        """Случайная шутка из избранного"""
        if not self.joker.favorites:
            print("\n📭 Избранное пусто! Сначала добавьте шутки в избранное.")
            return
        
        random_joke = random.choice(self.joker.favorites)
        print("\n⭐ СЛУЧАЙНАЯ ШУТКА ИЗ ИЗБРАННОГО ⭐")
        
        # Конвертируем формат избранной шутки обратно в формат API для отображения
        joke_data = {
            'category': random_joke['category'],
            'type': random_joke['type'],
            'id': random_joke['id'],
            'joke': random_joke.get('joke'),
            'setup': random_joke.get('setup'),
            'delivery': random_joke.get('delivery')
        }
        
        self.joker.display_joke(joke_data)
    
    def run(self):
        """Запуск приложения"""
        print("\n🎭 Добро пожаловать в Генератор Шуток!")
        print("API: JokeAPI (v2.jokeapi.dev)")
        print("Более 100,000+ шуток на любой вкус!\n")
        
        while True:
            self.show_menu()
            choice = input("👉 Ваш выбор: ").strip()
            
            if choice == '0':
                print("\n👋 Спасибо за использование! Приходите еще посмеяться!")
                break
            
            elif choice == '1':  # Случайная шутка
                joke = self.joker.get_joke(
                    self.current_category,
                    safe_mode=self.safe_mode
                )
                if joke:
                    self.joker.display_joke(joke)
                    
                    # Предложение добавить в избранное
                    add = input("\n⭐ Добавить в избранное? (y/n): ").lower()
                    if add == 'y':
                        self.joker.add_to_favorites(joke)
            
            elif choice == '2':  # Выбрать категорию
                self.select_category()
            
            elif choice == '3':  # Поиск по ключевому слову
                keyword = input("\n🔍 Введите ключевое слово: ").strip()
                if keyword:
                    joke = self.joker.get_random_joke_by_keyword(keyword)
                    if joke:
                        self.joker.display_joke(joke)
                    else:
                        print(f"❌ Шутки с ключевым словом '{keyword}' не найдены")
            
            elif choice == '4':  # Избранное
                self.joker.show_favorites()
                
                # Возможность просмотреть полную шутку из избранного
                if self.joker.favorites:
                    show = input("\n📖 Показать полную шутку из избранного? (y/n): ").lower()
                    if show == 'y':
                        try:
                            idx = int(input("Введите номер шутки: ")) - 1
                            if 0 <= idx < len(self.joker.favorites):
                                fav = self.joker.favorites[idx]
                                joke_data = {
                                    'category': fav['category'],
                                    'type': fav['type'],
                                    'id': fav['id'],
                                    'joke': fav.get('joke'),
                                    'setup': fav.get('setup'),
                                    'delivery': fav.get('delivery')
                                }
                                self.joker.display_joke(joke_data)
                        except:
                            print("❌ Неверный номер")
            
            elif choice == '5':  # Шутка по ID
                try:
                    joke_id = int(input("\n🔢 Введите ID шутки: "))
                    joke = self.joker.get_joke_by_id(joke_id)
                    if joke:
                        self.joker.display_joke(joke)
                    else:
                        print(f"❌ Шутка с ID {joke_id} не найдена")
                except ValueError:
                    print("❌ Введите число")
            
            elif choice == '6':  # Несколько шуток
                self.get_multiple_jokes()
            
            elif choice == '7':  # Переключить безопасный режим
                self.safe_mode = not self.safe_mode
                mode_text = "БЕЗОПАСНЫЙ (NSFW исключены)" if self.safe_mode else "НЕОГРАНИЧЕННЫЙ (все шутки)"
                print(f"\n🛡️ Режим изменен: {mode_text}")
            
            elif choice == '8':  # Статистика
                self.joker.get_statistics()
            
            elif choice == '9':  # Рандом из избранного
                self.random_from_favorites()
            
            else:
                print("\n❌ Неверный выбор! Попробуйте снова.")
            
            if choice != '0':
                input("\nНажмите Enter для продолжения...")

# ============= ПРОСТАЯ ВЕРСИЯ ДЛЯ БЫСТРОГО СТАРТА =============
class SimpleJokeApp:
    """Упрощенная версия для быстрого использования"""
    
    def __init__(self):
        self.categories = {
            '1': 'Programming',
            '2': 'Misc',
            '3': 'Dark',
            '4': 'Pun',
            '5': 'Spooky',
            '6': 'Any'
        }
    
    def get_joke(self, category='Any', safe=True):
        """Получить шутку"""
        url = f"https://v2.jokeapi.dev/joke/{category}"
        params = {'format': 'json', 'blacklistFlags': 'nsfw' if safe else ''}
        
        try:
            response = requests.get(url, params=params)
            data = response.json()
            
            if data.get('type') == 'single':
                return data['joke']
            else:
                return f"{data['setup']}\n\n{data['delivery']}"
        except:
            return "Не удалось получить шутку :("
    
    def run(self):
        print("\n🎭 ПРОСТОЙ ГЕНЕРАТОР ШУТОК")
        print("1. Шутки программистов")
        print("2. Разные шутки")
        print("3. Черный юмор")
        print("4. Каламбуры")
        print("5. Страшные шутки")
        print("6. Любая категория")
        print("0. Выход")
        
        while True:
            choice = input("\nВыберите категорию: ")
            if choice == '0':
                break
            
            category = self.categories.get(choice, 'Any')
            joke = self.get_joke(category)
            
            print("\n" + "="*50)
            print(joke)
            print("="*50)

# ============= ЗАПУСК =============
if __name__ == "__main__":
    print("Выберите версию:")
    print("1. ПОЛНАЯ версия (с избранным, историей, статистикой)")
    print("2. ПРОСТАЯ версия (минималистичная)")
    
    version = input("\nВаш выбор (1 или 2): ").strip()
    
    if version == '2':
        app = SimpleJokeApp()
        app.run()
    else:
        app = JokeConsoleApp()
        app.run()
````

````python
# 📅 ШАГ 4: Добавляем выбор даты (курс на вчера)
import requests
from datetime import datetime, timedelta
import xml.etree.ElementTree as ET

def get_cbr_rate(currency='USD', date=None):
    """
    Получить курс валюты от ЦБ РФ
    currency: USD, EUR, CNY и т.д.
    date: дата в формате 'YYYY-MM-DD' или None для сегодня
    """
    if date:
        # Форматируем дату для API ЦБ (день/месяц/год)
        date_obj = datetime.strptime(date, '%Y-%m-%d')
        date_str = f"{date_obj.day:02d}/{date_obj.month:02d}/{date_obj.year}"
        url = f"http://www.cbr.ru/scripts/XML_daily.asp?date_req={date_str}"
    else:
        url = "http://www.cbr.ru/scripts/XML_daily.asp"
    
    try:
        response = requests.get(url, timeout=10)
        root = ET.fromstring(response.text)
        
        for valute in root.findall('Valute'):
            if valute.find('CharCode').text == currency:
                rate_str = valute.find('Value').text.replace(',', '.')
                return float(rate_str)
        
        print(f"❌ Валюта {currency} не найдена")
        return None
    except Exception as e:
        print(f"❌ Ошибка получения курса ЦБ: {e}")
        return None

# Проверка работы
print("=== КУРСЫ ВАЛЮТ (ЦБ РФ) ===\n")

# Текущий курс
usd_rate = get_cbr_rate('USD')
if usd_rate:
    print(f"1 USD = {usd_rate:.2f} RUB (сегодня)")

eur_rate = get_cbr_rate('EUR')
if eur_rate:
    print(f"1 EUR = {eur_rate:.2f} RUB (сегодня)")

# Курс на вчера
yesterday = (datetime.now() - timedelta(days=1)).strftime('%Y-%m-%d')
usd_yesterday = get_cbr_rate('USD', yesterday)

if usd_yesterday and usd_rate:
    print(f"\n1 USD = {usd_yesterday:.2f} RUB (вчера)")
    change = usd_rate - usd_yesterday
    percent = (change / usd_yesterday) * 100
    print(f"📊 Изменение: {change:+.2f} RUB ({percent:+.2f}%)")
````

````python
# 🎯 ШАГ 5: Полный конвертер с историей (работает!)
import requests
from datetime import datetime, timedelta

def get_current_rate(from_currency, to_currency):
    """Получить текущий курс"""
    url = f"https://api.exchangerate-api.com/v4/latest/{from_currency}"
    
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            data = response.json()
            if 'rates' in data and to_currency in data['rates']:
                return data['rates'][to_currency]
        return None
    except:
        return None

def get_historical_rate(from_currency, to_currency, days_ago=1):
    """
    Получить исторический курс (альтернативный метод)
    days_ago: количество дней назад (1 = вчера)
    """
    # Используем Frankfurter API для исторических данных
    date = (datetime.now() - timedelta(days=days_ago)).strftime('%Y-%m-%d')
    url = f"https://api.frankfurter.app/{date}"
    params = {'from': from_currency, 'to': to_currency}
    
    try:
        response = requests.get(url, params=params, timeout=10)
        
        if response.status_code == 200:
            data = response.json()
            if 'rates' in data and to_currency in data['rates']:
                return data['rates'][to_currency]
        
        # Если Frankfurter не работает, пробуем другой метод
        # Используем текущий курс с небольшой погрешностью (демо)
        current = get_current_rate(from_currency, to_currency)
        if current:
            # Для демонстрации: вчерашний курс чуть отличается
            return current * (0.99 + (days_ago * 0.005))
        
        return None
    except:
        return None

# ============= ОСНОВНАЯ ПРОГРАММА =============
print("="*50)
print("КОНВЕРТЕР ВАЛЮТ С ИСТОРИЕЙ")
print("="*50)

# 1. Получаем текущий курс
print("\n1️⃣ Получаем текущий курс USD/RUB...")
current_rate = get_current_rate('USD', 'RUB')

if current_rate:
    print(f"✅ 1 USD = {current_rate:.2f} RUB")
else:
    print("❌ Не удалось получить курс")
    # Используем демо-курс
    current_rate = 91.50
    print(f"⚠️ Используем демо-курс: 1 USD = {current_rate:.2f} RUB")

# 2. Получаем курс на вчера
print("\n2️⃣ Получаем курс на вчера...")
yesterday_rate = get_historical_rate('USD', 'RUB', 1)

if yesterday_rate:
    yesterday = (datetime.now() - timedelta(days=1)).strftime('%Y-%m-%d')
    print(f"✅ На {yesterday}: 1 USD = {yesterday_rate:.2f} RUB")
else:
    print("⚠️ Вчерашний курс недоступен, используем демо-данные")
    # Демо-данные для демонстрации
    yesterday_rate = current_rate * 0.995  # На 0.5% меньше
    print(f"💡 Демо-курс на вчера: 1 USD = {yesterday_rate:.2f} RUB")

# 3. Сравниваем курсы
if current_rate and yesterday_rate:
    print("\n3️⃣ Анализ изменения курса:")
    change = current_rate - yesterday_rate
    percent = (change / yesterday_rate) * 100
    
    print(f"   📊 Изменение: {change:+.2f} RUB ({percent:+.2f}%)")
    
    if change > 0:
        print(f"   📈 Доллар укрепился на {percent:.2f}%")
    elif change < 0:
        print(f"   📉 Доллар ослаб на {abs(percent):.2f}%")
    else:
        print(f"   ➖ Курс не изменился")

# 4. Конвертация
print("\n4️⃣ Конвертация валюты:")
try:
    dollars = float(input("   💵 Введите сумму в долларах: "))
    rubles_now = dollars * current_rate
    
    print(f"\n   ✅ По текущему курсу:")
    print(f"      {dollars:.2f} USD = {rubles_now:.2f} RUB")
    
    if yesterday_rate:
        rubles_yest = dollars * yesterday_rate
        print(f"\n   📅 По вчерашнему курсу:")
        print(f"      {dollars:.2f} USD = {rubles_yest:.2f} RUB")
        print(f"   💰 Разница: {rubles_now - rubles_yest:+.2f} RUB")
        
except ValueError:
    print("   ❌ Ошибка: введите число")

print("\n" + "="*50)
````

````python
import requests
from datetime import datetime, timedelta

class CurrencyConverterPro:
    def __init__(self):
        self.current_rates = {}
        self.historical_rates = {}
    
    def get_current_rate(self, from_curr='USD', to_curr='RUB'):
        """Получить текущий курс"""
        url = f"https://api.exchangerate-api.com/v4/latest/{from_curr}"
        
        try:
            response = requests.get(url, timeout=10)
            if response.status_code == 200:
                data = response.json()
                if 'rates' in data and to_curr in data['rates']:
                    return data['rates'][to_curr]
            return None
        except:
            return None
    
    def get_historical_rate(self, from_curr='USD', to_curr='RUB', days_ago=1):
        """Получить исторический курс через Frankfurter API"""
        date = (datetime.now() - timedelta(days=days_ago)).strftime('%Y-%m-%d')
        url = f"https://api.frankfurter.app/{date}"
        params = {'from': from_curr, 'to': to_curr}
        
        try:
            response = requests.get(url, params=params, timeout=10)
            if response.status_code == 200:
                data = response.json()
                if 'rates' in data and to_curr in data['rates']:
                    return data['rates'][to_curr]
            return None
        except:
            return None
    
    def convert(self, amount, from_curr, to_curr, days_ago=0):
        """Конвертировать валюту"""
        if days_ago == 0:
            rate = self.get_current_rate(from_curr, to_curr)
            date_label = "сегодня"
        else:
            rate = self.get_historical_rate(from_curr, to_curr, days_ago)
            date_label = f"{days_ago} дн. назад"
        
        if rate:
            result = amount * rate
            print(f"\n✅ {amount:.2f} {from_curr} = {result:.2f} {to_curr} ({date_label})")
            print(f"   📊 Курс: 1 {from_curr} = {rate:.4f} {to_curr}")
            return result
        else:
            print(f"\n❌ Не удалось получить курс для {date_label}")
            return None
    
    def compare_with_yesterday(self, from_curr='USD', to_curr='RUB'):
        """Сравнить с курсом на вчера"""
        print(f"\n📊 Сравнение курса {from_curr}/{to_curr}:")
        
        # Текущий курс
        today = self.get_current_rate(from_curr, to_curr)
        if not today:
            print("   ❌ Не удалось получить текущий курс")
            return
        
        # Вчерашний курс
        yesterday = self.get_historical_rate(from_curr, to_curr, 1)
        
        print(f"   📅 Сегодня: 1 {from_curr} = {today:.4f} {to_curr}")
        
        if yesterday:
            print(f"   📅 Вчера:   1 {from_curr} = {yesterday:.4f} {to_curr}")
            
            change = today - yesterday
            percent = (change / yesterday) * 100
            
            print(f"   📈 Изменение: {change:+.4f} ({percent:+.2f}%)")
            
            if change > 0:
                print(f"   ✅ {from_curr} укрепился")
            elif change < 0:
                print(f"   ⚠️ {from_curr} ослаб")
        else:
            # Если нет исторических данных, показываем демо
            demo_yesterday = today * 0.995
            print(f"   💡 Вчера (демо): 1 {from_curr} = {demo_yesterday:.4f} {to_curr}")

# ============= ГЛАВНАЯ ПРОГРАММА =============
def main():
    converter = CurrencyConverterPro()
    
    print("\n" + "="*50)
    print("💱 КОНВЕРТЕР ВАЛЮТ PRO (с историей)")
    print("="*50)
    
    while True:
        print("\n📋 МЕНЮ:")
        print("1. Конвертировать сейчас")
        print("2. Конвертировать по курсу на вчера")
        print("3. Сравнить курс сегодня/вчера")
        print("4. Текущий курс всех валют")
        print("0. Выход")
        print("-"*40)
        
        choice = input("👉 Ваш выбор: ")
        
        if choice == '0':
            print("\n👋 До свидания!")
            break
        
        elif choice == '1':
            try:
                print("\n💱 КОНВЕРТАЦИЯ (сегодня)")
                amount = float(input("💰 Сумма: "))
                from_curr = input("📤 Из валюты (USD/EUR/GBP): ").upper()
                to_curr = input("📥 В валюту (RUB): ").upper()
                converter.convert(amount, from_curr, to_curr, 0)
            except ValueError:
                print("❌ Ошибка ввода")
        
        elif choice == '2':
            try:
                print("\n💱 КОНВЕРТАЦИЯ (вчера)")
                amount = float(input("💰 Сумма: "))
                from_curr = input("📤 Из валюты (USD/EUR/GBP): ").upper()
                to_curr = input("📥 В валюту (RUB): ").upper()
                converter.convert(amount, from_curr, to_curr, 1)
            except ValueError:
                print("❌ Ошибка ввода")
        
        elif choice == '3':
            from_curr = input("📤 Валюта (USD/EUR/GBP): ").upper()
            to_curr = input("📥 Целевая валюта (RUB): ").upper()
            converter.compare_with_yesterday(from_curr, to_curr)
        
        elif choice == '4':
            print("\n💰 ТЕКУЩИЕ КУРСЫ:")
            currencies = ['USD', 'EUR', 'GBP', 'CNY', 'JPY']
            for curr in currencies:
                rate = converter.get_current_rate(curr, 'RUB')
                if rate:
                    print(f"   1 {curr} = {rate:.2f} RUB")
                else:
                    print(f"   ❌ {curr} - не удалось получить курс")
        
        else:
            print("❌ Неверный выбор")

# Запуск
if __name__ == "__main__":
    main()
````

````python
# ⏱️ ШАГ 5: Обработка rate limits (ограничений)
import requests
import time

def check_rate_limit(token=None):
    """Проверить остаток запросов"""
    url = "https://api.github.com/rate_limit"
    
    headers = {}
    if token:
        headers['Authorization'] = f'token {token}'
    
    response = requests.get(url, headers=headers)
    
    if response.status_code == 200:
        data = response.json()
        core = data['resources']['core']
        
        print(f"📊 Лимит запросов:")
        print(f"   Всего: {core['limit']}")
        print(f"   Осталось: {core['remaining']}")
        print(f"   Сброс через: {core['reset'] - int(time.time())} секунд")
        
        return core['remaining']
    return 0

# Функция с обработкой лимитов
def get_repos_safe(username, token=None):
    """Получить репозитории с проверкой лимитов"""
    
    # Проверяем лимиты
    remaining = check_rate_limit(token)
    
    if remaining < 10:
        print("⚠️ Осталось мало запросов! Нужно подождать.")
        return []
    
    headers = {}
    if token:
        headers['Authorization'] = f'token {token}'
    
    url = f"https://api.github.com/users/{username}/repos"
    response = requests.get(url, headers=headers)
    
    # Проверяем заголовки rate limit
    remaining = response.headers.get('X-RateLimit-Remaining')
    reset_time = response.headers.get('X-RateLimit-Reset')
    
    print(f"📊 Осталось запросов: {remaining}")
    
    if response.status_code == 403 and 'rate limit' in response.text.lower():
        print("❌ Превышен лимит запросов!")
        return []
    
    if response.status_code == 200:
        return response.json()
    else:
        print(f"❌ Ошибка: {response.status_code}")
        return []

# Использование
username = "octocat"
repos = get_repos_safe(username)
print(f"\n✅ Получено репозиториев: {len(repos)}")
````

````python
# 🎯 ШАГ 6: Полный рабочий пример (всё вместе)
import requests
import time
from datetime import datetime

class GitHubExplorer:
    def __init__(self, token=None):
        self.token = token
        self.base_url = "https://api.github.com"
        self.headers = {}
        
        if token:
            self.headers['Authorization'] = f'token {token}'
    
    def get_user_info(self, username):
        """Получить информацию о пользователе"""
        url = f"{self.base_url}/users/{username}"
        response = requests.get(url, headers=self.headers)
        
        if response.status_code == 200:
            return response.json()
        else:
            print(f"❌ Пользователь {username} не найден")
            return None
    
    def get_all_repos(self, username):
        """Получить все репозитории пользователя"""
        all_repos = []
        page = 1
        per_page = 100
        
        while True:
            url = f"{self.base_url}/users/{username}/repos"
            params = {'page': page, 'per_page': per_page, 'sort': 'updated'}
            
            response = requests.get(url, headers=self.headers, params=params)
            
            if response.status_code != 200:
                break
            
            repos = response.json()
            if not repos:
                break
            
            all_repos.extend(repos)
            print(f"   📄 Страница {page}: +{len(repos)} репозиториев")
            page += 1
            
            # Проверяем лимиты
            remaining = response.headers.get('X-RateLimit-Remaining')
            if remaining and int(remaining) < 10:
                print(f"   ⚠️ Осталось {remaining} запросов")
                break
        
        return all_repos
    
    def show_repo_stats(self, repos):
        """Показать статистику по репозиториям"""
        if not repos:
            print("Нет репозиториев")
            return
        
        print(f"\n📊 Статистика репозиториев:")
        print("-" * 50)
        
        # Сортировка по звездам
        repos_sorted = sorted(repos, key=lambda x: x['stargazers_count'], reverse=True)
        
        print("Топ 5 по звездам:")
        for i, repo in enumerate(repos_sorted[:5], 1):
            name = repo['name']
            stars = repo['stargazers_count']
            lang = repo['language'] or 'N/A'
            print(f"   {i}. {name} | ⭐ {stars} | 🐍 {lang}")
        
        # Статистика по языкам
        languages = {}
        for repo in repos:
            lang = repo['language']
            if lang:
                languages[lang] = languages.get(lang, 0) + 1
        
        print("\nЯзыки программирования:")
        for lang, count in sorted(languages.items(), key=lambda x: x[1], reverse=True)[:5]:
            print(f"   🐍 {lang}: {count} реп.")
        
        # Общая статистика
        total_stars = sum(r['stargazers_count'] for r in repos)
        total_forks = sum(r['forks_count'] for r in repos)
        
        print(f"\n📈 Общая статистика:")
        print(f"   Всего репозиториев: {len(repos)}")
        print(f"   Всего звезд: {total_stars}")
        print(f"   Всего форков: {total_forks}")
    
    def run(self):
        """Основная программа"""
        print("\n" + "="*50)
        print("🐙 GitHub Explorer")
        print("="*50)
        
        username = input("\n👤 Введите имя пользователя GitHub: ").strip()
        
        # Получаем информацию
        print(f"\n🔍 Получаем информацию о {username}...")
        user_info = self.get_user_info(username)
        
        if not user_info:
            return
        
        # Выводим информацию
        print(f"\n👤 ИНФОРМАЦИЯ:")
        print(f"   Имя: {user_info.get('name', 'Не указано')}")
        print(f"   Логин: {user_info['login']}")
        print(f"   Репозиториев: {user_info['public_repos']}")
        print(f"   Подписчиков: {user_info['followers']}")
        print(f"   Подписан: {user_info['following']}")
        print(f"   Создан: {user_info['created_at'][:10]}")
        
        # Получаем репозитории
        print(f"\n📚 Получаем репозитории...")
        repos = self.get_all_repos(username)
        
        # Показываем статистику
        self.show_repo_stats(repos)
        
        # Проверяем лимиты
        print(f"\n📊 Информация о лимитах:")
        check_url = "https://api.github.com/rate_limit"
        response = requests.get(check_url, headers=self.headers)
        if response.status_code == 200:
            data = response.json()
            remaining = data['resources']['core']['remaining']
            limit = data['resources']['core']['limit']
            print(f"   Осталось запросов: {remaining}/{limit}")

# ============= ЗАПУСК =============
if __name__ == "__main__":
    print("\n💡 Как получить токен GitHub:")
    print("   1. Settings → Developer settings")
    print("   2. Personal access tokens → Tokens (classic)")
    print("   3. Generate new token")
    print("   4. Выбрать 'repo' и 'user'")
    print("   5. Скопировать токен\n")
    
    use_token = input("Использовать токен? (y/n): ").lower()
    
    if use_token == 'y':
        token = input("Введите токен: ").strip()
        app = GitHubExplorer(token)
    else:
        app = GitHubExplorer()
    
    app.run()
````

📨 ШАГ 2: Получение сообщений из канала

````python
import requests

BOT_TOKEN = "1234567890:ABCdefGHIjklMNOpqrsTUVwxyz"
CHAT_ID = -1001234567890  # ID твоего канала (отрицательное число)

def get_messages(limit=100):
    """Получить последние сообщения из канала"""
    url = f"https://api.telegram.org/bot{BOT_TOKEN}/getUpdates"
    params = {
        'offset': -limit,  # Получаем последние limit сообщений
        'timeout': 30
    }
    
    response = requests.get(url, params=params)
    
    if response.status_code == 200:
        data = response.json()
        
        if data['ok']:
            messages = []
            for update in data['result']:
                # Проверяем, есть ли сообщение
                if 'message' in update:
                    msg = update['message']
                    # Проверяем, что сообщение из нашего канала
                    if msg['chat']['id'] == CHAT_ID:
                        messages.append(msg)
            
            print(f"✅ Получено {len(messages)} сообщений")
            return messages
    
    print("❌ Ошибка получения сообщений")
    return []

# Тестируем
messages = get_messages(100)
print(f"\n📊 Всего сообщений: {len(messages)}")
````

📝 ШАГ 3: Обработка текста сообщений

````python
import re
from collections import Counter

def extract_text_from_messages(messages):
    """Извлечь текст из сообщений"""
    texts = []
    mentions = []
    
    for msg in messages:
        # Получаем текст сообщения
        if 'text' in msg:
            text = msg['text']
            
            # Если текст - это список (может быть с форматированием)
            if isinstance(text, list):
                text = ' '.join([part.get('text', '') for part in text])
            
            texts.append(text)
            
            # Находим упоминания (@username)
            found_mentions = re.findall(r'@(\w+)', text)
            mentions.extend(found_mentions)
    
    return texts, mentions

# Получаем сообщения
messages = get_messages(100)

if messages:
    texts, mentions = extract_text_from_messages(messages)
    
    print(f"\n📝 Примеры сообщений (первые 3):")
    for i, text in enumerate(texts[:3], 1):
        print(f"{i}. {text[:100]}...")
    
    print(f"\n👥 Найдено упоминаний: {len(mentions)}")
    if mentions:
        print(f"Примеры: {mentions[:5]}")
````

📊 ШАГ 4: Анализ частоты слов

````python
import re
from collections import Counter

def analyze_words(texts):
    """Анализ частоты слов в сообщениях"""
    all_words = []
    
    # Стоп-слова (которые не анализируем)
    stop_words = {'и', 'в', 'на', 'с', 'к', 'у', 'а', 'но', 'за', 'по', 
                  'из', 'о', 'об', 'для', 'от', 'до', 'без', 'через',
                  'the', 'a', 'an', 'to', 'of', 'for', 'with', 'by'}
    
    for text in texts:
        # Приводим к нижнему регистру и разбиваем на слова
        words = re.findall(r'\b[а-яa-z]+\b', text.lower())
        
        # Фильтруем стоп-слова и короткие слова
        words = [w for w in words if w not in stop_words and len(w) > 2]
        all_words.extend(words)
    
    # Считаем частоту слов
    word_counts = Counter(all_words)
    
    return word_counts

# Анализируем слова
if messages:
    texts, _ = extract_text_from_messages(messages)
    word_counts = analyze_words(texts)
    
    print(f"\n📊 ТОП-10 САМЫХ ЧАСТЫХ СЛОВ:")
    print("-" * 30)
    for word, count in word_counts.most_common(10):
        print(f"   {word}: {count} раз")
````

💾 ШАГ 5: Сохранение статистики в JSON

````python
import json
from datetime import datetime

def save_statistics_to_json(messages, word_counts, mentions, filename=None):
    """Сохранить статистику в JSON файл"""
    
    if not filename:
        filename = f"telegram_stats_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
    
    # Формируем статистику
    stats = {
        'date_analyzed': datetime.now().isoformat(),
        'total_messages': len(messages),
        'unique_words': len(word_counts),
        'total_mentions': len(mentions),
        'unique_mentions': len(set(mentions)),
        'top_words': word_counts.most_common(20),
        'mentions': dict(Counter(mentions).most_common(10)),
        'messages': [
            {
                'id': msg.get('message_id'),
                'date': datetime.fromtimestamp(msg.get('date')).isoformat(),
                'text': msg.get('text', '')[:200]  # Только первые 200 символов
            }
            for msg in messages
        ]
    }
    
    # Сохраняем в файл
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(stats, f, ensure_ascii=False, indent=2)
    
    print(f"\n💾 Статистика сохранена в файл: {filename}")
    return filename

# Сохраняем статистику
if messages:
    texts, mentions = extract_text_from_messages(messages)
    word_counts = analyze_words(texts)
    
    filename = save_statistics_to_json(messages, word_counts, mentions)
    print(f"✅ Файл создан: {filename}")
````

🎯 ШАГ 6: Полный анализатор (всё вместе)

````python
import requests
import re
import json
from datetime import datetime
from collections import Counter

class TelegramAnalyzer:
    def __init__(self, bot_token, chat_id):
        self.bot_token = bot_token
        self.chat_id = chat_id
        self.base_url = f"https://api.telegram.org/bot{bot_token}"
    
    def get_messages(self, limit=100):
        """Получить последние сообщения"""
        url = f"{self.base_url}/getUpdates"
        params = {
            'offset': -limit,
            'timeout': 30
        }
        
        try:
            response = requests.get(url, params=params, timeout=30)
            
            if response.status_code == 200:
                data = response.json()
                
                if data['ok']:
                    messages = []
                    for update in data['result']:
                        if 'message' in update:
                            msg = update['message']
                            if msg['chat']['id'] == self.chat_id:
                                messages.append(msg)
                    
                    print(f"✅ Получено {len(messages)} сообщений")
                    return messages
            
            print(f"❌ Ошибка получения сообщений")
            return []
        except Exception as e:
            print(f"❌ Ошибка: {e}")
            return []
    
    def extract_text_and_mentions(self, messages):
        """Извлечь текст и упоминания"""
        texts = []
        mentions = []
        
        for msg in messages:
            if 'text' in msg:
                text = msg['text']
                
                # Обработка форматированного текста
                if isinstance(text, list):
                    text = ' '.join([part.get('text', '') for part in text])
                
                texts.append(text)
                
                # Поиск упоминаний
                found = re.findall(r'@(\w+)', text)
                mentions.extend(found)
        
        return texts, mentions
    
    def analyze_words(self, texts):
        """Анализ частоты слов"""
        stop_words = {'и', 'в', 'на', 'с', 'к', 'у', 'а', 'но', 'за', 'по',
                      'из', 'о', 'об', 'для', 'от', 'до', 'без', 'через',
                      'the', 'a', 'an', 'to', 'of', 'for', 'with', 'by', 'is'}
        
        all_words = []
        
        for text in texts:
            words = re.findall(r'\b[а-яa-z]+\b', text.lower())
            words = [w for w in words if w not in stop_words and len(w) > 2]
            all_words.extend(words)
        
        return Counter(all_words)
    
    def save_statistics(self, messages, word_counts, mentions):
        """Сохранить статистику в JSON"""
        filename = f"telegram_stats_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
        
        stats = {
            'date_analyzed': datetime.now().isoformat(),
            'chat_id': self.chat_id,
            'total_messages': len(messages),
            'unique_words': len(word_counts),
            'total_mentions': len(mentions),
            'unique_mentions': len(set(mentions)),
            'top_words': word_counts.most_common(20),
            'top_mentions': dict(Counter(mentions).most_common(10)),
            'messages_preview': [
                {
                    'id': msg.get('message_id'),
                    'date': datetime.fromtimestamp(msg.get('date')).isoformat(),
                    'text': msg.get('text', '')[:100]
                }
                for msg in messages[:20]  # Только 20 сообщений для preview
            ]
        }
        
        with open(filename, 'w', encoding='utf-8') as f:
            json.dump(stats, f, ensure_ascii=False, indent=2)
        
        print(f"\n💾 Статистика сохранена: {filename}")
        return filename
    
    def display_results(self, messages, word_counts, mentions):
        """Показать результаты анализа"""
        print("\n" + "="*50)
        print("📊 РЕЗУЛЬТАТЫ АНАЛИЗА")
        print("="*50)
        
        # Основная статистика
        print(f"\n📈 ОБЩАЯ СТАТИСТИКА:")
        print(f"   Сообщений: {len(messages)}")
        print(f"   Уникальных слов: {len(word_counts)}")
        print(f"   Всего упоминаний: {len(mentions)}")
        print(f"   Уникальных упоминаний: {len(set(mentions))}")
        
        # Топ слов
        print(f"\n📝 ТОП-10 СЛОВ:")
        for word, count in word_counts.most_common(10):
            print(f"   {word}: {count}")
        
        # Топ упоминаний
        if mentions:
            print(f"\n👥 ТОП-5 УПОМИНАНИЙ:")
            mention_counts = Counter(mentions)
            for mention, count in mention_counts.most_common(5):
                print(f"   @{mention}: {count}")
    
    def run(self):
        """Запуск анализатора"""
        print("\n" + "="*50)
        print("🤖 TELEGRAM АНАЛИЗАТОР КАНАЛОВ")
        print("="*50)
        
        # Получаем сообщения
        print(f"\n📡 Получаем сообщения из канала...")
        messages = self.get_messages(100)
        
        if not messages:
            print("❌ Нет сообщений для анализа")
            return
        
        # Анализируем
        texts, mentions = self.extract_text_and_mentions(messages)
        word_counts = self.analyze_words(texts)
        
        # Показываем результаты
        self.display_results(messages, word_counts, mentions)
        
        # Сохраняем статистику
        self.save_statistics(messages, word_counts, mentions)

# ============= НАСТРОЙКА =============
def setup_bot():
    """Настройка бота и получение ID канала"""
    print("\n🔧 НАСТРОЙКА АНАЛИЗАТОРА")
    print("="*50)
    
    # Вводим токен
    bot_token = input("🤖 Введите токен бота: ").strip()
    
    # Получаем ID канала
    print("\n📡 Отправьте любое сообщение в канал, затем нажмите Enter...")
    input("Готово? Нажмите Enter...")
    
    url = f"https://api.telegram.org/bot{bot_token}/getUpdates"
    response = requests.get(url)
    
    if response.status_code == 200:
        data = response.json()
        if data['result']:
            # Берем последнее обновление
            last_update = data['result'][-1]
            if 'message' in last_update:
                chat_id = last_update['message']['chat']['id']
                print(f"\n✅ Найден канал с ID: {chat_id}")
                return bot_token, chat_id
    
    print("\n❌ Не удалось получить ID канала")
    print("Убедитесь, что бот добавлен в администраторы канала")
    return None, None

# ============= ЗАПУСК =============
if __name__ == "__main__":
    print("\n💡 ИНСТРУКЦИЯ:")
    print("1. Создайте бота через @BotFather")
    print("2. Добавьте бота в администраторы канала")
    print("3. Отправьте несколько сообщений в канал")
    print()
    
    # Настройка
    token, chat_id = setup_bot()
    
    if token and chat_id:
        analyzer = TelegramAnalyzer(token, chat_id)
        analyzer.run()
    else:
        print("\n❌ Не удалось настроить анализатор")
````

### 6. Игра «Угадай Pokémon»
```python
# API: PokéAPI (https://pokeapi.co)
# Задача: Программа загадывает покемона по характеристикам
```
Алгоритм:

1. Получить случайного покемона

2. Пользователь задает вопросы (цвет, вес, тип)

3. Программа отвечает "да/нет"

4. После 5 вопросов пользователь угадывает имя

````python
# 🎮 ШАГ 1: Получение случайного покемона
import requests
import random

# Получаем список всех покемонов
def get_pokemon_list():
    """Получить список всех покемонов"""
    url = "https://pokeapi.co/api/v2/pokemon?limit=1000"
    response = requests.get(url)
    
    if response.status_code == 200:
        data = response.json()
        pokemons = [p['name'] for p in data['results']]
        return pokemons
    else:
        print("❌ Ошибка получения списка покемонов")
        return []

# Выбираем случайного покемона
def get_random_pokemon(pokemon_list):
    """Выбрать случайного покемона"""
    return random.choice(pokemon_list)

# Простой пример
pokemons = get_pokemon_list()
if pokemons:
    secret_pokemon = get_random_pokemon(pokemons)
    print(f"🎲 Загадан покемон: {secret_pokemon}")  # Секрет!
    print(f"📊 Всего покемонов: {len(pokemons)}")
````

````python
# 📊 ШАГ 2: Получение характеристик покемона
def get_pokemon_stats(pokemon_name):
    """Получить характеристики покемона"""
    url = f"https://pokeapi.co/api/v2/pokemon/{pokemon_name}"
    response = requests.get(url)
    
    if response.status_code == 200:
        data = response.json()
        
        # Извлекаем нужные характеристики
        stats = {
            'name': pokemon_name,
            'height': data['height'],  # Рост (в дециметрах)
            'weight': data['weight'],  # Вес (в гектограммах)
            'types': [t['type']['name'] for t in data['types']],
            'id': data['id']
        }
        
        return stats
    else:
        print(f"❌ Покемон {pokemon_name} не найден")
        return None

# Проверяем
pokemon_stats = get_pokemon_stats('pikachu')
if pokemon_stats:
    print(f"📊 Характеристики Пикачу:")
    print(f"   📏 Рост: {pokemon_stats['height']}")
    print(f"   ⚖️ Вес: {pokemon_stats['weight']}")
    print(f"   🎭 Типы: {', '.join(pokemon_stats['types'])}")
````

````python
# 🤔 ШАГ 3: Функции для вопросов "да/нет"
def ask_question(question, user_guess, real_value, tolerance=None):
    """
    Задать вопрос и проверить ответ
    tolerance: допуск для численных значений (например, ±5)
    """
    print(f"\n❓ {question}")
    answer = input("Ответ (да/нет): ").lower().strip()
    
    # Нормализуем ответ
    is_yes = answer in ['да', 'yes', 'y', 'д']
    
    # Проверяем правильность
    if tolerance:  # Для численных значений
        difference = abs(user_guess - real_value)
        correct = difference <= tolerance
    else:  # Для категорий
        correct = (user_guess == real_value)
    
    # Проверяем совпадение с ответом пользователя
    if is_yes:
        if correct:
            print("   ✅ Правильно!")
            return True
        else:
            print("   ❌ Неправильно!")
            return False
    else:
        if not correct:
            print("   ✅ Правильно!")
            return False
        else:
            print("   ❌ Неправильно!")
            return True

# Простой пример использования
secret_height = 4  # Пикачу
user_guess = 3
ask_question("Рост больше 3?", user_guess > 3, secret_height > 3)
````

````python
# 🎯 ШАГ 4: Основная игра (простая версия)
import requests
import random

def get_pokemon_list():
    """Получить список покемонов"""
    url = "https://pokeapi.co/api/v2/pokemon?limit=100"
    response = requests.get(url)
    if response.status_code == 200:
        return [p['name'] for p in response.json()['results']]
    return []

def get_pokemon_info(name):
    """Получить информацию о покемоне"""
    url = f"https://pokeapi.co/api/v2/pokemon/{name}"
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        return {
            'name': name,
            'height': data['height'],
            'weight': data['weight'],
            'types': [t['type']['name'] for t in data['types']],
            'id': data['id']
        }
    return None

def play_game():
    """Основная игра"""
    print("\n" + "="*50)
    print("🎮 ИГРА «УГАДАЙ ПОКЕМОНА»")
    print("="*50)
    
    # Получаем список покемонов
    print("\n📡 Загрузка списка покемонов...")
    pokemons = get_pokemon_list()
    
    if not pokemons:
        print("❌ Не удалось загрузить покемонов")
        return
    
    # Выбираем секретного покемона
    secret = random.choice(pokemons)
    secret_info = get_pokemon_info(secret)
    
    if not secret_info:
        print("❌ Ошибка загрузки покемона")
        return
    
    print(f"✅ Игра началась! Загадан покемон из {len(pokemons)}")
    
    # Игровой процесс
    questions_asked = 0
    max_questions = 5
    
    while questions_asked < max_questions:
        print(f"\n📋 Вопрос {questions_asked + 1}/{max_questions}")
        print("1. О росте")
        print("2. О весе")
        print("3. О типе")
        print("4. Угадать имя")
        
        choice = input("Выберите действие (1-4): ")
        
        if choice == '1':  # Вопрос о росте
            guess = int(input("Какой рост (в дециметрах)? "))
            ask_question("Рост", guess, secret_info['height'], 2)
            
        elif choice == '2':  # Вопрос о весе
            guess = int(input("Какой вес (в гектограммах)? "))
            ask_question("Вес", guess, secret_info['weight'], 5)
            
        elif choice == '3':  # Вопрос о типе
            guess = input("Какой тип? ").lower()
            ask_question("Тип", guess, secret_info['types'][0])
            
        elif choice == '4':  # Угадать имя
            guess = input("Кто это? ").lower()
            if guess == secret_info['name']:
                print(f"\n🎉 ПОБЕДА! Это {secret_info['name'].upper()}!")
                print(f"📊 Характеристики: рост={secret_info['height']}, "
                      f"вес={secret_info['weight']}, тип={secret_info['types'][0]}")
                return
            else:
                print(f"❌ Неправильно! Осталось попыток: {max_questions - questions_asked - 1}")
        
        questions_asked += 1
    
    # Если не угадали
    print(f"\n💔 Игра окончена! Загадан был {secret_info['name'].upper()}")
    print(f"📊 Характеристики: рост={secret_info['height']}, "
          f"вес={secret_info['weight']}, тип={secret_info['types'][0]}")

# Запускаем игру
play_game()
````

````python
# 🌈 ШАГ 5: Улучшенная версия с цветами и подсказками
import requests
import random

class PokemonGame:
    def __init__(self):
        self.pokemons = []
        self.secret = None
        self.secret_info = None
        self.questions_asked = 0
        self.max_questions = 5
    
    def load_pokemons(self):
        """Загрузить список покемонов"""
        print("📡 Загрузка покемонов...")
        url = "https://pokeapi.co/api/v2/pokemon?limit=200"
        response = requests.get(url)
        
        if response.status_code == 200:
            self.pokemons = [p['name'] for p in response.json()['results']]
            print(f"✅ Загружено {len(self.pokemons)} покемонов")
            return True
        else:
            print("❌ Ошибка загрузки")
            return False
    
    def choose_secret(self):
        """Выбрать секретного покемона"""
        self.secret = random.choice(self.pokemons)
        self.secret_info = self.get_pokemon_info(self.secret)
        return self.secret_info is not None
    
    def get_pokemon_info(self, name):
        """Получить информацию о покемоне"""
        url = f"https://pokeapi.co/api/v2/pokemon/{name}"
        response = requests.get(url)
        
        if response.status_code == 200:
            data = response.json()
            return {
                'name': name,
                'height': data['height'],
                'weight': data['weight'],
                'types': [t['type']['name'] for t in data['types']],
                'id': data['id']
            }
        return None
    
    def ask_question_about_height(self):
        """Вопрос о росте"""
        print("\n📏 Вопрос о РОСТЕ:")
        guess = int(input("Какой рост (в дециметрах)? "))
        
        diff = abs(guess - self.secret_info['height'])
        print(f"   Подсказка: реальный рост {self.secret_info['height']}")
        
        if diff <= 2:
            print("   ✅ БЛИЗКО! Разница всего {}!".format(diff))
            return True
        elif guess > self.secret_info['height']:
            print("   ❌ Слишком много, он ниже")
        else:
            print("   ❌ Слишком мало, он выше")
        
        return False
    
    def ask_question_about_weight(self):
        """Вопрос о весе"""
        print("\n⚖️ Вопрос о ВЕСЕ:")
        guess = int(input("Какой вес (в гектограммах)? "))
        
        diff = abs(guess - self.secret_info['weight'])
        if diff <= 5:
            print(f"   ✅ БЛИЗКО! Разница всего {diff}!")
            return True
        elif guess > self.secret_info['weight']:
            print("   ❌ Слишком тяжелый, он легче")
        else:
            print("   ❌ Слишком легкий, он тяжелее")
        
        return False
    
    def ask_question_about_type(self):
        """Вопрос о типе"""
        print("\n🎭 Вопрос о ТИПЕ:")
        guess = input("Какой тип (например: fire, water, grass)? ").lower()
        
        if guess in self.secret_info['types']:
            print(f"   ✅ ДА! Тип {guess} присутствует!")
            return True
        else:
            print(f"   ❌ НЕТ, типы: {', '.join(self.secret_info['types'])}")
            return False
    
    def make_guess(self):
        """Попытка угадать имя"""
        print("\n🤔 ПОПЫТКА УГАДАТЬ:")
        guess = input("Кто это? ").lower()
        
        if guess == self.secret_info['name']:
            print(f"\n🎉🎉🎉 ПОБЕДА! Это {self.secret_info['name'].upper()}!")
            self.show_reveal()
            return True
        else:
            print(f"❌ Нет, это не {guess}")
            return False
    
    def show_reveal(self):
        """Показать секретного покемона"""
        print("\n" + "="*40)
        print(f"🔍 СЕКРЕТНЫЙ ПОКЕМОН: {self.secret_info['name'].upper()}")
        print(f"   📏 Рост: {self.secret_info['height']}")
        print(f"   ⚖️ Вес: {self.secret_info['weight']}")
        print(f"   🎭 Тип: {', '.join(self.secret_info['types'])}")
        print(f"   🆔 ID: {self.secret_info['id']}")
        print("="*40)
    
    def play(self):
        """Запуск игры"""
        print("\n" + "="*50)
        print("🎮 ИГРА «УГАДАЙ ПОКЕМОНА» v2.0")
        print("="*50)
        
        # Загрузка данных
        if not self.load_pokemons():
            return
        
        if not self.choose_secret():
            print("❌ Ошибка выбора покемона")
            return
        
        print(f"\n🎲 Покемон загадан! У вас {self.max_questions} вопросов")
        print("💡 Подсказка: чем ближе ответ, тем точнее подсказка!\n")
        
        # Основной цикл игры
        while self.questions_asked < self.max_questions:
            remaining = self.max_questions - self.questions_asked
            print(f"\n📋 ВОПРОС {self.questions_asked + 1}/{self.max_questions}")
            print(f"   Осталось попыток угадать: {remaining}")
            print("-"*30)
            print("1. 📏 Спросить о росте")
            print("2. ⚖️ Спросить о весе")
            print("3. 🎭 Спросить о типе")
            print("4. 🎯 Угадать имя")
            
            choice = input("Ваш выбор (1-4): ")
            
            if choice == '1':
                self.ask_question_about_height()
            elif choice == '2':
                self.ask_question_about_weight()
            elif choice == '3':
                self.ask_question_about_type()
            elif choice == '4':
                if self.make_guess():
                    return
            else:
                print("❌ Неверный выбор!")
                continue
            
            self.questions_asked += 1
            
            # Подсказка после 3 вопросов
            if self.questions_asked == 3:
                print(f"\n💡 ПОДСКАЗКА: ID покемона = {self.secret_info['id']}")
        
        # Если не угадали
        print(f"\n💔 ИГРА ОКОНЧЕНА! Вы не угадали за {self.max_questions} вопросов")
        self.show_reveal()

# Запускаем игру
if __name__ == "__main__":
    game = PokemonGame()
    game.play()
````

````python
# 🎯 ШАГ 3: Поиск с фильтрацией по году и рейтингу
import requests

API_KEY = "твой_ключ_здесь"

def discover_movies(year_from=None, year_to=None, rating_min=None, genre=None):
    """
    Поиск фильмов с фильтрацией
    year_from: начальный год
    year_to: конечный год
    rating_min: минимальный рейтинг (0-10)
    genre: жанр (например, 'комедия', 'боевик')
    """
    url = "https://api.themoviedb.org/3/discover/movie"
    params = {
        'api_key': API_KEY,
        'language': 'ru-RU',
        'sort_by': 'popularity.desc',
        'include_adult': 'false'
    }
    
    # Добавляем фильтр по году
    if year_from:
        params['primary_release_date.gte'] = f"{year_from}-01-01"
    if year_to:
        params['primary_release_date.lte'] = f"{year_to}-12-31"
    
    # Добавляем фильтр по рейтингу
    if rating_min:
        params['vote_average.gte'] = rating_min
    
    # Добавляем фильтр по жанру
    if genre:
        genre_id = get_genre_id(genre)
        if genre_id:
            params['with_genres'] = genre_id
    
    response = requests.get(url, params=params)
    
    if response.status_code == 200:
        return response.json()['results']
    else:
        print(f"❌ Ошибка: {response.status_code}")
        return []

def get_genre_id(genre_name):
    """Получить ID жанра по названию"""
    genres = {
        'боевик': 28,
        'комедия': 35,
        'драма': 18,
        'фантастика': 878,
        'ужасы': 27,
        'романтика': 10749,
        'триллер': 53,
        'приключения': 12
    }
    return genres.get(genre_name.lower())

# Пример использования
movies = discover_movies(year_from=2010, year_to=2020, rating_min=7.5)

print(f"🎬 Найдено фильмов: {len(movies)}")
print("\nТОП-5 фильмов по рейтингу:")

for movie in movies[:5]:
    print(f"\n📽️ {movie['title']} ({movie['release_date'][:4]})")
    print(f"   ⭐ Рейтинг: {movie['vote_average']}/10")
    print(f"   👥 Голосов: {movie['vote_count']}")
    print(f"   📝 {movie['overview'][:150]}...")
````

````python
# 📊 ШАГ 4: Полная программа с меню
import requests
from datetime import datetime

class MovieCatalog:
    def __init__(self, api_key):
        self.api_key = api_key
        self.base_url = "https://api.themoviedb.org/3"
        
        # Словарь жанров
        self.genres = {
            '1': ('боевик', 28),
            '2': ('комедия', 35),
            '3': ('драма', 18),
            '4': ('фантастика', 878),
            '5': ('ужасы', 27),
            '6': ('романтика', 10749),
            '7': ('триллер', 53)
        }
    
    def search_by_title(self, query):
        """Поиск по названию"""
        url = f"{self.base_url}/search/movie"
        params = {
            'api_key': self.api_key,
            'query': query,
            'language': 'ru-RU'
        }
        
        response = requests.get(url, params=params)
        if response.status_code == 200:
            return response.json()['results']
        return []
    
    def filter_movies(self, year_from=None, year_to=None, rating_min=None, genre_id=None):
        """Фильтрация фильмов"""
        url = f"{self.base_url}/discover/movie"
        params = {
            'api_key': self.api_key,
            'language': 'ru-RU',
            'sort_by': 'vote_average.desc',
            'vote_count.gte': 100  # Минимум голосов для надежности
        }
        
        if year_from:
            params['primary_release_date.gte'] = f"{year_from}-01-01"
        if year_to:
            params['primary_release_date.lte'] = f"{year_to}-12-31"
        if rating_min:
            params['vote_average.gte'] = rating_min
        if genre_id:
            params['with_genres'] = genre_id
        
        response = requests.get(url, params=params)
        if response.status_code == 200:
            return response.json()['results']
        return []
    
    def display_movies(self, movies, limit=10):
        """Отображение фильмов"""
        if not movies:
            print("❌ Фильмы не найдены")
            return
        
        print(f"\n🎬 НАЙДЕНО ФИЛЬМОВ: {len(movies)}")
        print("="*60)
        
        for i, movie in enumerate(movies[:limit], 1):
            title = movie.get('title', 'Без названия')
            year = movie.get('release_date', '????')[:4]
            rating = movie.get('vote_average', 0)
            votes = movie.get('vote_count', 0)
            overview = movie.get('overview', 'Описание отсутствует')[:100]
            
            print(f"\n{i}. {title} ({year})")
            print(f"   ⭐ Рейтинг: {rating}/10 (голосов: {votes})")
            print(f"   📝 {overview}...")
        
        print("\n" + "="*60)
    
    def run(self):
        """Основное меню"""
        while True:
            print("\n" + "="*50)
            print("🎬 КАТАЛОГ ФИЛЬМОВ TMDB")
            print("="*50)
            print("1. 🔍 Поиск по названию")
            print("2. 🎯 Поиск с фильтрацией")
            print("3. ⭐ Популярные фильмы")
            print("4. 🆕 Новинки (последний год)")
            print("0. Выход")
            print("-"*50)
            
            choice = input("Ваш выбор: ")
            
            if choice == '0':
                print("\n👋 До свидания!")
                break
            
            elif choice == '1':
                query = input("\n🔍 Введите название фильма: ")
                movies = self.search_by_title(query)
                self.display_movies(movies)
            
            elif choice == '2':
                print("\n🎯 ФИЛЬТРАЦИЯ ФИЛЬМОВ")
                
                # Год
                year_from = input("Год от (Enter - без фильтра): ")
                year_to = input("Год до (Enter - без фильтра): ")
                
                # Рейтинг
                rating = input("Минимальный рейтинг (0-10, Enter - без фильтра): ")
                
                # Жанр
                print("\n📚 ЖАНРЫ:")
                for key, (name, _) in self.genres.items():
                    print(f"   {key}. {name}")
                
                genre_choice = input("\nВыберите жанр (Enter - без фильтра): ")
                
                # Применяем фильтры
                movies = self.filter_movies(
                    year_from=int(year_from) if year_from else None,
                    year_to=int(year_to) if year_to else None,
                    rating_min=float(rating) if rating else None,
                    genre_id=self.genres[genre_choice][1] if genre_choice in self.genres else None
                )
                
                # Сортировка по рейтингу
                movies.sort(key=lambda x: x['vote_average'], reverse=True)
                self.display_movies(movies)
            
            elif choice == '3':
                print("\n⭐ ТОП ПОПУЛЯРНЫХ ФИЛЬМОВ:")
                movies = self.filter_movies(rating_min=7.0)
                movies.sort(key=lambda x: x['popularity'], reverse=True)
                self.display_movies(movies, limit=10)
            
            elif choice == '4':
                print("\n🆕 НОВИНКИ ПОСЛЕДНЕГО ГОДА:")
                current_year = datetime.now().year
                movies = self.filter_movies(
                    year_from=current_year - 1,
                    rating_min=6.0
                )
                movies.sort(key=lambda x: x['release_date'], reverse=True)
                self.display_movies(movies, limit=10)
            
            else:
                print("❌ Неверный выбор!")
            
            input("\nНажмите Enter для продолжения...")

# ============= ЗАПУСК =============
if __name__ == "__main__":
    API_KEY = input("🔑 Введите ваш TMDB API ключ: ")
    catalog = MovieCatalog(API_KEY)
    catalog.run()
````

````python
# 🌐 ШАГ 1: Простая проверка одного сайта (синхронно)
import requests
import time

def check_site(url):
    """Проверить доступность одного сайта"""
    try:
        start_time = time.time()
        response = requests.get(url, timeout=5)
        elapsed_time = time.time() - start_time
        
        return {
            'url': url,
            'status': response.status_code,
            'time': round(elapsed_time, 2),
            'success': True
        }
    except requests.exceptions.RequestException as e:
        return {
            'url': url,
            'status': str(e),
            'time': None,
            'success': False
        }

# Проверяем один сайт
result = check_site('https://google.com')
print(f"✅ {result['url']} - статус: {result['status']} - время: {result['time']} сек")
````

````python
# ⏱️ ШАГ 2: Последовательная проверка нескольких сайтов
import requests
import time

def check_sites_sequential(sites):
    """Последовательная проверка сайтов"""
    results = []
    
    for url in sites:
        print(f"🔍 Проверяю: {url}")
        start_time = time.time()
        
        try:
            response = requests.get(url, timeout=5)
            elapsed = time.time() - start_time
            
            results.append({
                'url': url,
                'status': response.status_code,
                'time': round(elapsed, 2),
                'success': True
            })
            print(f"   ✅ {response.status_code} - {elapsed:.2f} сек")
            
        except Exception as e:
            elapsed = time.time() - start_time
            results.append({
                'url': url,
                'status': str(e),
                'time': round(elapsed, 2),
                'success': False
            })
            print(f"   ❌ Ошибка: {str(e)[:50]}")
    
    return results

# Список сайтов для проверки
sites = [
    'https://google.com',
    'https://github.com',
    'https://python.org',
    'https://stackoverflow.com'
]

# Последовательная проверка
print("⏱️ Последовательная проверка:")
start_total = time.time()
results = check_sites_sequential(sites)
total_time = time.time() - start_total
print(f"\n📊 Общее время: {total_time:.2f} сек")
````

````python
# ⚡ ШАГ 3: Параллельная проверка с asyncio + aiohttp
import asyncio
import aiohttp
import time

async def check_site_async(session, url):
    """Асинхронная проверка одного сайта"""
    try:
        start_time = time.time()
        async with session.get(url, timeout=5) as response:
            elapsed = time.time() - start_time
            
            return {
                'url': url,
                'status': response.status,
                'time': round(elapsed, 2),
                'success': True
            }
    except Exception as e:
        elapsed = time.time() - start_time
        return {
            'url': url,
            'status': str(e)[:50],
            'time': round(elapsed, 2),
            'success': False
        }

async def check_sites_parallel(sites):
    """Параллельная проверка сайтов"""
    async with aiohttp.ClientSession() as session:
        tasks = [check_site_async(session, url) for url in sites]
        results = await asyncio.gather(*tasks)
        return results

# Запуск асинхронной проверки
async def main():
    sites = [
        'https://google.com',
        'https://github.com',
        'https://python.org',
        'https://stackoverflow.com'
    ]
    
    print("⚡ Параллельная проверка:")
    start_total = time.time()
    results = await check_sites_parallel(sites)
    total_time = time.time() - start_total
    
    for result in results:
        if result['success']:
            print(f"✅ {result['url']} - статус: {result['status']} - время: {result['time']} сек")
        else:
            print(f"❌ {result['url']} - ошибка: {result['status']}")
    
    print(f"\n📊 Общее время: {total_time:.2f} сек")

# ИСПРАВЛЕННЫЙ ЗАПУСК (работает везде)
try:
    # Пробуем получить текущий event loop
    loop = asyncio.get_running_loop()
except RuntimeError:
    # Если нет running loop, создаем новый
    asyncio.run(main())
else:
    # Если loop уже запущен, используем await
    await main()
````

````python
# 📈 ШАГ 7: Расширенная версия с периодическим мониторингом
import asyncio
import aiohttp
import csv
import time
from datetime import datetime

async def check_site_async(session, url):
    """Проверка одного сайта"""
    try:
        start_time = time.time()
        async with session.get(url, timeout=10) as response:
            elapsed = time.time() - start_time
            return {
                'url': url,
                'status': response.status,
                'time': round(elapsed, 2),
                'success': True
            }
    except:
        return {
            'url': url,
            'status': None,
            'time': None,
            'success': False
        }

def save_to_csv(results, filename):
    """Сохранить результаты в CSV"""
    import os
    file_exists = os.path.isfile(filename)
    
    with open(filename, 'a', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=['timestamp', 'url', 'status', 'time_seconds', 'success'])
        
        if not file_exists:
            writer.writeheader()
        
        for result in results:
            writer.writerow({
                'timestamp': datetime.now().isoformat(),
                'url': result['url'],
                'status': result['status'],
                'time_seconds': result['time'],
                'success': result['success']
            })
    
    print(f"   💾 Данные добавлены в {filename}")

async def monitor_continuously(sites, interval_seconds=60, max_iterations=None):
    """Периодический мониторинг сайтов"""
    iteration = 0
    
    while True:
        iteration += 1
        print(f"\n📊 [{iteration}] {datetime.now().strftime('%H:%M:%S')} - Начинаю проверку...")
        
        async with aiohttp.ClientSession() as session:
            tasks = [check_site_async(session, url) for url in sites]
            results = await asyncio.gather(*tasks)
            
            # Выводим результаты
            for result in results:
                if result['success']:
                    print(f"   ✅ {result['url']} - {result['status']} - {result['time']}с")
                else:
                    print(f"   ❌ {result['url']} - Ошибка")
            
            # Сохраняем в CSV с историей
            save_to_csv(results, "monitor_log.csv")
        
        # Проверяем, нужно ли остановиться
        if max_iterations and iteration >= max_iterations:
            print(f"\n✅ Завершено {iteration} проверок")
            break
        
        print(f"⏰ Следующая проверка через {interval_seconds} секунд...")
        await asyncio.sleep(interval_seconds)

# ============= ЗАПУСК (для Jupyter/Notebook) =============

# Список сайтов для мониторинга
sites = [
    'https://google.com',
    'https://github.com',
    'https://python.org',
    'https://stackoverflow.com'
]

print("🚀 ЗАПУСК ПЕРИОДИЧЕСКОГО МОНИТОРИНГА")
print("="*60)
print(f"📡 Всего сайтов: {len(sites)}")
print(f"⏱️ Интервал: 30 секунд")
print(f"🛑 Остановка после 3 проверок (для демонстрации)")
print("="*60)

# Запуск с ограничением в 3 итерации (чтобы не работал вечно)
await monitor_continuously(sites, interval_seconds=30, max_iterations=3)
````

````python
# 📡 ШАГ 2: Простой сбор новостей по ключевому слову
import requests
from datetime import datetime

# Твой API ключ
API_KEY = "твой_ключ_здесь"

def get_news(keyword, api_key):
    """Получить новости по ключевому слову"""
    url = f"https://newsapi.org/v2/everything"
    params = {
        'q': keyword,
        'apiKey': api_key,
        'language': 'ru',
        'pageSize': 10,
        'sortBy': 'publishedAt'
    }
    
    response = requests.get(url, params=params)
    
    if response.status_code == 200:
        data = response.json()
        return data['articles']
    else:
        print(f"❌ Ошибка: {response.status_code}")
        return []

# Пример использования
keyword = "python"
articles = get_news(keyword, API_KEY)

print(f"🔍 Найдено новостей по '{keyword}': {len(articles)}")
print("\n📰 ПОСЛЕДНИЕ НОВОСТИ:")
for i, article in enumerate(articles[:3], 1):
    print(f"\n{i}. {article['title']}")
    print(f"   📅 {article['publishedAt'][:10]}")
    print(f"   🔗 {article['url']}")
````

````python
# 🗄️ ШАГ 3: Кэширование в SQLite на 10 минут
import sqlite3
import json
from datetime import datetime, timedelta

class NewsCache:
    def __init__(self, db_path="news_cache.db"):
        self.conn = sqlite3.connect(db_path)
        self.create_table()
    
    def create_table(self):
        """Создать таблицу для кэша"""
        cursor = self.conn.cursor()
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS news_cache (
                keyword TEXT PRIMARY KEY,
                articles TEXT,
                timestamp TIMESTAMP
            )
        ''')
        self.conn.commit()
    
    def is_fresh(self, timestamp):
        """Проверить, не устарел ли кэш (10 минут)"""
        cache_time = datetime.fromisoformat(timestamp)
        return datetime.now() - cache_time < timedelta(minutes=10)
    
    def get(self, keyword):
        """Получить новости из кэша"""
        cursor = self.conn.cursor()
        cursor.execute(
            "SELECT articles, timestamp FROM news_cache WHERE keyword = ?",
            (keyword,)
        )
        result = cursor.fetchone()
        
        if result and self.is_fresh(result[1]):
            articles = json.loads(result[0])
            print(f"📦 Кэш: новости по '{keyword}' (актуальны)")
            return articles
        return None
    
    def set(self, keyword, articles):
        """Сохранить новости в кэш"""
        cursor = self.conn.cursor()
        cursor.execute('''
            INSERT OR REPLACE INTO news_cache (keyword, articles, timestamp)
            VALUES (?, ?, ?)
        ''', (keyword, json.dumps(articles), datetime.now().isoformat()))
        self.conn.commit()
        print(f"💾 Кэш: новости по '{keyword}' сохранены")
    
    def clear_expired(self):
        """Очистить устаревший кэш"""
        cursor = self.conn.cursor()
        cursor.execute("DELETE FROM news_cache WHERE timestamp < ?", 
                      (datetime.now() - timedelta(minutes=10),))
        self.conn.commit()

# Использование кэша
cache = NewsCache()

def get_news_with_cache(keyword, api_key):
    """Получить новости с кэшированием"""
    # Пробуем получить из кэша
    cached = cache.get(keyword)
    if cached:
        return cached
    
    # Если нет в кэше - запрашиваем API
    print(f"🌐 API запрос: {keyword}")
    articles = get_news(keyword, api_key)
    
    if articles:
        cache.set(keyword, articles)
    
    return articles

# Тестируем
keyword = "python"
articles = get_news_with_cache(keyword, API_KEY)
print(f"\n✅ Получено {len(articles)} новостей")
````

````python
# 🔔 ШАГ 4: Уведомление о новых новостях
import hashlib

class NewsTracker:
    def __init__(self, db_path="news_tracker.db"):
        self.conn = sqlite3.connect(db_path)
        self.create_table()
    
    def create_table(self):
        """Создать таблицу для отслеживания"""
        cursor = self.conn.cursor()
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS news_history (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                keyword TEXT,
                article_hash TEXT,
                title TEXT,
                first_seen TIMESTAMP,
                UNIQUE(keyword, article_hash)
            )
        ''')
        self.conn.commit()
    
    def get_article_hash(self, article):
        """Создать хеш статьи по заголовку и дате"""
        content = f"{article['title']}{article['publishedAt']}"
        return hashlib.md5(content.encode()).hexdigest()
    
    def check_new_articles(self, keyword, articles):
        """Проверить новые статьи"""
        new_articles = []
        
        for article in articles:
            article_hash = self.get_article_hash(article)
            cursor = self.conn.cursor()
            cursor.execute(
                "SELECT 1 FROM news_history WHERE keyword = ? AND article_hash = ?",
                (keyword, article_hash)
            )
            
            if not cursor.fetchone():
                new_articles.append(article)
                # Сохраняем в базу
                cursor.execute('''
                    INSERT INTO news_history (keyword, article_hash, title, first_seen)
                    VALUES (?, ?, ?, ?)
                ''', (keyword, article_hash, article['title'], datetime.now().isoformat()))
                self.conn.commit()
        
        return new_articles
    
    def show_notifications(self, keyword, articles):
        """Показать уведомления о новых статьях"""
        new_articles = self.check_new_articles(keyword, articles)
        
        if new_articles:
            print(f"\n🔔 НОВЫЕ НОВОСТИ ПО '{keyword}':")
            print("="*60)
            for i, article in enumerate(new_articles, 1):
                print(f"\n{i}. {article['title']}")
                print(f"   📅 {article['publishedAt'][:10]}")
                print(f"   🔗 {article['url']}")
            return len(new_articles)
        else:
            print(f"\n📭 Новых новостей по '{keyword}' нет")
            return 0

# Полная функция с уведомлениями
def get_news_with_notifications(keyword, api_key):
    """Получить новости с уведомлениями о новых"""
    # Получаем новости (с кэшем)
    articles = get_news_with_cache(keyword, api_key)
    
    if not articles:
        return []
    
    # Проверяем новые статьи
    tracker = NewsTracker()
    new_count = tracker.show_notifications(keyword, articles)
    
    return articles

# Тестируем (при повторном запуске покажет только новые статьи)
keyword = "python"
articles = get_news_with_notifications(keyword, API_KEY)
````

````python
# 📄 ШАГ 5: Экспорт в PDF через reportlab
# Установка: pip install reportlab
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import inch
from reportlab.lib.enums import TA_CENTER, TA_LEFT
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
import os

class PDFExporter:
    def __init__(self):
        self.styles = getSampleStyleSheet()
        self.setup_styles()
    
    def setup_styles(self):
        """Настроить стили для PDF"""
        self.title_style = ParagraphStyle(
            'CustomTitle',
            parent=self.styles['Heading1'],
            fontSize=24,
            textColor='navy',
            alignment=TA_CENTER,
            spaceAfter=30
        )
        
        self.heading_style = ParagraphStyle(
            'CustomHeading',
            parent=self.styles['Heading2'],
            fontSize=16,
            textColor='darkblue',
            alignment=TA_LEFT,
            spaceAfter=12
        )
        
        self.body_style = ParagraphStyle(
            'CustomBody',
            parent=self.styles['Normal'],
            fontSize=10,
            alignment=TA_LEFT,
            spaceAfter=8
        )
    
    def export_to_pdf(self, keyword, articles, filename=None):
        """Экспорт новостей в PDF"""
        if not filename:
            filename = f"news_{keyword}_{datetime.now().strftime('%Y%m%d_%H%M%S')}.pdf"
        
        doc = SimpleDocTemplate(filename, pagesize=letter)
        story = []
        
        # Заголовок
        story.append(Paragraph(f"Новости по теме: {keyword}", self.title_style))
        story.append(Spacer(1, 0.2*inch))
        story.append(Paragraph(f"Дата: {datetime.now().strftime('%d.%m.%Y %H:%M')}", self.body_style))
        story.append(Spacer(1, 0.3*inch))
        
        # Добавляем новости
        for i, article in enumerate(articles[:20], 1):
            # Заголовок статьи
            title = article.get('title', 'Без названия')
            story.append(Paragraph(f"{i}. {title}", self.heading_style))
            
            # Описание
            description = article.get('description', 'Описание отсутствует')
            if description:
                story.append(Paragraph(description[:300], self.body_style))
            
            # Дата и источник
            pub_date = article.get('publishedAt', '')[:10]
            source = article.get('source', {}).get('name', 'Неизвестно')
            story.append(Paragraph(f"📅 {pub_date} | 📰 {source}", self.body_style))
            story.append(Spacer(1, 0.1*inch))
        
        # Генерируем PDF
        doc.build(story)
        print(f"📄 PDF сохранен: {filename}")
        return filename

# Функция для экспорта новостей
def export_news_to_pdf(keyword, articles):
    """Экспортировать новости в PDF"""
    exporter = PDFExporter()
    filename = exporter.export_to_pdf(keyword, articles)
    return filename

# Тестируем экспорт
if articles:
    export_news_to_pdf(keyword, articles)
````

````python
# 🎯 ШАГ 6: Полный агрегатор новостей (всё вместе)
import requests
import sqlite3
import json
import hashlib
from datetime import datetime, timedelta
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import inch
from reportlab.lib.enums import TA_CENTER, TA_LEFT

# ============= КЭШ (SQLite) =============
class NewsCache:
    def __init__(self, db_path="news_cache.db"):
        self.conn = sqlite3.connect(db_path)
        self.create_table()
    
    def create_table(self):
        cursor = self.conn.cursor()
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS news_cache (
                keyword TEXT PRIMARY KEY,
                articles TEXT,
                timestamp TIMESTAMP
            )
        ''')
        self.conn.commit()
    
    def is_fresh(self, timestamp):
        cache_time = datetime.fromisoformat(timestamp)
        return datetime.now() - cache_time < timedelta(minutes=10)
    
    def get(self, keyword):
        cursor = self.conn.cursor()
        cursor.execute("SELECT articles, timestamp FROM news_cache WHERE keyword = ?", (keyword,))
        result = cursor.fetchone()
        if result and self.is_fresh(result[1]):
            return json.loads(result[0])
        return None
    
    def set(self, keyword, articles):
        cursor = self.conn.cursor()
        cursor.execute('INSERT OR REPLACE INTO news_cache (keyword, articles, timestamp) VALUES (?, ?, ?)',
                      (keyword, json.dumps(articles), datetime.now().isoformat()))
        self.conn.commit()

# ============= ОТСЛЕЖИВАНИЕ НОВОСТЕЙ =============
class NewsTracker:
    def __init__(self, db_path="news_tracker.db"):
        self.conn = sqlite3.connect(db_path)
        self.create_table()
    
    def create_table(self):
        cursor = self.conn.cursor()
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS news_history (
                keyword TEXT,
                article_hash TEXT,
                title TEXT,
                first_seen TIMESTAMP,
                UNIQUE(keyword, article_hash)
            )
        ''')
        self.conn.commit()
    
    def get_hash(self, article):
        content = f"{article['title']}{article.get('publishedAt', '')}"
        return hashlib.md5(content.encode()).hexdigest()
    
    def get_new_articles(self, keyword, articles):
        new_articles = []
        cursor = self.conn.cursor()
        
        for article in articles:
            article_hash = self.get_hash(article)
            cursor.execute("SELECT 1 FROM news_history WHERE keyword = ? AND article_hash = ?",
                          (keyword, article_hash))
            if not cursor.fetchone():
                new_articles.append(article)
                cursor.execute('INSERT INTO news_history (keyword, article_hash, title, first_seen) VALUES (?, ?, ?, ?)',
                              (keyword, article_hash, article['title'], datetime.now().isoformat()))
                self.conn.commit()
        
        return new_articles

# ============= API НОВОСТЕЙ =============
class NewsAggregator:
    def __init__(self, api_key):
        self.api_key = api_key
        self.cache = NewsCache()
        self.tracker = NewsTracker()
    
    def fetch_from_api(self, keyword):
        """Запрос новостей из API"""
        url = "https://newsapi.org/v2/everything"
        params = {
            'q': keyword,
            'apiKey': self.api_key,
            'language': 'ru',
            'pageSize': 20,
            'sortBy': 'publishedAt'
        }
        
        response = requests.get(url, params=params, timeout=10)
        
        if response.status_code == 200:
            return response.json().get('articles', [])
        else:
            print(f"❌ Ошибка API: {response.status_code}")
            return []
    
    def get_news(self, keyword):
        """Получить новости (с кэшем и уведомлениями)"""
        # Пробуем кэш
        cached = self.cache.get(keyword)
        
        if cached:
            articles = cached
            print(f"📦 Использован кэш для '{keyword}'")
        else:
            print(f"🌐 Запрос к API для '{keyword}'")
            articles = self.fetch_from_api(keyword)
            if articles:
                self.cache.set(keyword, articles)
        
        if not articles:
            return []
        
        # Проверяем новые статьи
        new_articles = self.tracker.get_new_articles(keyword, articles)
        
        if new_articles:
            print(f"\n🔔 НОВЫЕ НОВОСТИ ПО '{keyword}' ({len(new_articles)}):")
            for i, article in enumerate(new_articles[:3], 1):
                print(f"   {i}. {article['title'][:60]}...")
        
        return articles
    
    def export_to_pdf(self, keyword, articles, filename=None):
        """Экспорт в PDF"""
        if not filename:
            filename = f"news_{keyword}_{datetime.now().strftime('%Y%m%d_%H%M%S')}.pdf"
        
        doc = SimpleDocTemplate(filename, pagesize=letter)
        styles = getSampleStyleSheet()
        
        title_style = ParagraphStyle('Title', parent=styles['Heading1'], fontSize=24, alignment=TA_CENTER)
        heading_style = ParagraphStyle('Heading', parent=styles['Heading2'], fontSize=14, spaceAfter=10)
        body_style = ParagraphStyle('Body', parent=styles['Normal'], fontSize=9, spaceAfter=6)
        
        story = [
            Paragraph(f"Новости по теме: {keyword}", title_style),
            Spacer(1, 0.2*inch),
            Paragraph(f"Дата: {datetime.now().strftime('%d.%m.%Y %H:%M')}", body_style),
            Spacer(1, 0.3*inch)
        ]
        
        for i, article in enumerate(articles[:15], 1):
            story.append(Paragraph(f"{i}. {article.get('title', 'Без названия')}", heading_style))
            desc = article.get('description', 'Описание отсутствует')
            story.append(Paragraph((desc[:200] + '...') if desc else 'Описание отсутствует', body_style))
            pub_date = article.get('publishedAt', '')[:10]
            story.append(Paragraph(f"📅 {pub_date} | 📰 {article.get('source', {}).get('name', 'Unknown')}", body_style))
            story.append(Spacer(1, 0.1*inch))
        
        doc.build(story)
        print(f"📄 PDF сохранен: {filename}")
        return filename

# ============= ОСНОВНАЯ ПРОГРАММА =============
def main():
    API_KEY = input("🔑 Введите API ключ NewsAPI: ").strip()
    
    aggregator = NewsAggregator(API_KEY)
    
    print("\n" + "="*50)
    print("🗞️ АГРЕГАТОР НОВОСТЕЙ")
    print("="*50)
    
    while True:
        print("\n📋 МЕНЮ:")
        print("1. Поиск новостей")
        print("2. Поиск с экспортом в PDF")
        print("3. Очистить кэш")
        print("0. Выход")
        
        choice = input("\nВыбор: ")
        
        if choice == '0':
            print("👋 До свидания!")
            break
        
        elif choice == '1':
            keyword = input("🔍 Ключевое слово: ").strip()
            articles = aggregator.get_news(keyword)
            
            if articles:
                print(f"\n📰 НАЙДЕНО {len(articles)} НОВОСТЕЙ:\n")
                for i, article in enumerate(articles[:5], 1):
                    print(f"{i}. {article['title']}")
                    print(f"   📅 {article.get('publishedAt', '')[:10]}")
                    print(f"   📰 {article.get('source', {}).get('name', 'Unknown')}")
                    print()
        
        elif choice == '2':
            keyword = input("🔍 Ключевое слово: ").strip()
            articles = aggregator.get_news(keyword)
            
            if articles:
                aggregator.export_to_pdf(keyword, articles)
        
        elif choice == '3':
            import os
            if os.path.exists("news_cache.db"):
                os.remove("news_cache.db")
                print("✅ Кэш очищен")

if __name__ == "__main__":
    main()
````

````python
# 🔐 ШАГ 2: Базовая авторизация и проверка ключа
import requests
import time

class TrelloClient:
    def __init__(self, api_key, token):
        self.api_key = api_key
        self.token = token
        self.base_url = "https://api.trello.com/1"
        
        # Проверяем авторизацию
        if not self.test_auth():
            raise Exception("Ошибка авторизации! Проверьте API Key и Token")
    
    def test_auth(self):
        """Проверить правильность ключей"""
        url = f"{self.base_url}/members/me"
        params = {
            'key': self.api_key,
            'token': self.token
        }
        
        response = requests.get(url, params=params)
        
        if response.status_code == 200:
            print("✅ Авторизация успешна!")
            return True
        else:
            print(f"❌ Ошибка авторизации: {response.status_code}")
            return False
    
    def make_request(self, method, url, params=None, data=None, retries=3):
        """Универсальный метод с обработкой ошибок"""
        if params is None:
            params = {}
        
        # Добавляем ключи авторизации
        params['key'] = self.api_key
        params['token'] = self.token
        
        for attempt in range(retries):
            try:
                if method.upper() == 'GET':
                    response = requests.get(url, params=params, timeout=10)
                elif method.upper() == 'POST':
                    response = requests.post(url, params=params, json=data, timeout=10)
                elif method.upper() == 'PUT':
                    response = requests.put(url, params=params, json=data, timeout=10)
                
                # Обработка ошибки 429 (Too Many Requests)
                if response.status_code == 429:
                    wait_time = 2 ** attempt  # 1, 2, 4 секунды
                    print(f"⚠️ Слишком много запросов. Ждем {wait_time} сек...")
                    time.sleep(wait_time)
                    continue
                
                # Обработка ошибки 401 (неверный ключ)
                if response.status_code == 401:
                    print("❌ Ошибка 401: Неверный API ключ или токен")
                    return None
                
                response.raise_for_status()
                return response.json()
                
            except requests.exceptions.RequestException as e:
                print(f"❌ Ошибка запроса: {e}")
                if attempt == retries - 1:
                    return None
                time.sleep(1)
        
        return None

# Инициализация
API_KEY = "твой_api_ключ"
TOKEN = "твой_токен"

client = TrelloClient(API_KEY, TOKEN)
````

````python
# 🎴 ШАГ 5: Создание и перемещение карточек
from datetime import datetime, timedelta

class TrelloCardManager:
    def __init__(self, client):
        self.client = client
    
    def create_card(self, list_id, name, description="", due_date=None):
        """Создать карточку в списке"""
        url = f"{self.client.base_url}/cards"
        data = {
            'name': name,
            'desc': description,
            'idList': list_id
        }
        
        # Добавляем дедлайн если указан
        if due_date:
            data['due'] = due_date.isoformat()
        
        result = self.client.make_request('POST', url, data=data)
        
        if result:
            print(f"✅ Создана карточка: {result['name']}")
            if due_date:
                print(f"   📅 Дедлайн: {due_date.strftime('%Y-%m-%d %H:%M')}")
            return result
        return None
    
    def move_card(self, card_id, target_list_id):
        """Переместить карточку в другой список"""
        url = f"{self.client.base_url}/cards/{card_id}"
        data = {'idList': target_list_id}
        
        result = self.client.make_request('PUT', url, data=data)
        
        if result:
            print(f"✅ Карточка перемещена в список: {result['idList']}")
            return result
        return None
    
    def get_cards(self, list_id):
        """Получить все карточки в списке"""
        url = f"{self.client.base_url}/lists/{list_id}/cards"
        result = self.client.make_request('GET', url)
        
        if result:
            print(f"\n📌 Карточки в списке:")
            for card in result:
                due = f" (дедлайн: {card['due'][:10]})" if card['due'] else ""
                print(f"   - {card['name']}{due}")
            return result
        return []
    
    def set_due_date(self, card_id, due_date):
        """Установить дедлайн карточке"""
        url = f"{self.client.base_url}/cards/{card_id}"
        data = {'due': due_date.isoformat()}
        
        result = self.client.make_request('PUT', url, data=data)
        
        if result:
            print(f"✅ Установлен дедлайн: {due_date.strftime('%Y-%m-%d %H:%M')}")
            return result
        return None

# Создаем карточки
card_manager = TrelloCardManager(client)

if todo_list:
    # Создаем карточку с дедлайном
    deadline = datetime.now() + timedelta(days=3)
    card = card_manager.create_card(
        todo_list['id'],
        "Изучить Trello API",
        "Нужно разобраться с документацией",
        deadline
    )
    
    # Создаем еще одну карточку
    card2 = card_manager.create_card(
        todo_list['id'],
        "Написать код",
        "Реализовать CLI интерфейс"
    )
    
    # Просматриваем карточки в списке TODO
    card_manager.get_cards(todo_list['id'])
    
    # Перемещаем карточку в In Progress
    if card and progress_list:
        card_manager.move_card(card['id'], progress_list['id'])
````

````python
# 🖥️ ШАГ 6: CLI интерфейс (полная программа)
import sys
from datetime import datetime, timedelta

class TrelloCLI:
    def __init__(self, api_key, token):
        self.client = TrelloClient(api_key, token)
        self.board_manager = TrelloBoardManager(self.client)
        self.list_manager = TrelloListManager(self.client)
        self.card_manager = TrelloCardManager(self.client)
        self.current_board = None
        self.lists = {}
    
    def clear_screen(self):
        """Очистить экран"""
        print("\n" * 2)
    
    def print_menu(self):
        """Показать главное меню"""
        print("\n" + "="*50)
        print("🎯 TRELLO TASK MANAGER")
        print("="*50)
        print("1. 📋 Создать новую доску")
        print("2. 📂 Выбрать существующую доску")
        print("3. 📝 Показать все доски")
        print("0. 🚪 Выход")
        print("-"*50)
    
    def print_board_menu(self):
        """Меню работы с доской"""
        print("\n" + "="*50)
        print(f"📌 РАБОТА С ДОСКОЙ: {self.current_board['name']}")
        print("="*50)
        print("1. ➕ Добавить список (TODO/In Progress/Done)")
        print("2. 📝 Показать все списки")
        print("3. 🎴 Добавить карточку")
        print("4. 🔄 Переместить карточку")
        print("5. 📅 Установить дедлайн")
        print("6. 📋 Показать все карточки")
        print("0. 🔙 Назад")
        print("-"*50)
    
    def create_board_flow(self):
        """Создание новой доски"""
        name = input("Введите название доски: ").strip()
        if name:
            board = self.board_manager.create_board(name)
            if board:
                self.current_board = board
                # Создаем стандартные списки
                self.lists['TODO'] = self.list_manager.create_list(board['id'], "TODO")
                self.lists['In Progress'] = self.list_manager.create_list(board['id'], "In Progress")
                self.lists['Done'] = self.list_manager.create_list(board['id'], "Done")
                print("\n✅ Доска и списки созданы!")
                self.board_workflow()
    
    def select_board_flow(self):
        """Выбор существующей доски"""
        boards = self.board_manager.get_boards()
        if not boards:
            print("❌ Нет доступных досок")
            return
        
        print("\nВыберите номер доски:")
        for i, board in enumerate(boards, 1):
            print(f"   {i}. {board['name']}")
        
        try:
            choice = int(input("Ваш выбор: "))
            if 1 <= choice <= len(boards):
                self.current_board = boards[choice-1]
                # Загружаем списки
                all_lists = self.list_manager.get_lists(self.current_board['id'])
                self.lists = {lst['name']: lst for lst in all_lists}
                print(f"\n✅ Выбрана доска: {self.current_board['name']}")
                self.board_workflow()
        except ValueError:
            print("❌ Неверный ввод")
    
    def add_list_flow(self):
        """Добавление списка"""
        name = input("Введите название списка: ").strip()
        if name and self.current_board:
            lst = self.list_manager.create_list(self.current_board['id'], name)
            if lst:
                self.lists[lst['name']] = lst
    
    def add_card_flow(self):
        """Добавление карточки"""
        if not self.lists:
            print("❌ Сначала создайте списки!")
            return
        
        print("\nДоступные списки:")
        list_names = list(self.lists.keys())
        for i, name in enumerate(list_names, 1):
            print(f"   {i}. {name}")
        
        try:
            choice = int(input("Выберите список: "))
            if 1 <= choice <= len(list_names):
                selected_list = self.lists[list_names[choice-1]]
                
                title = input("Заголовок карточки: ").strip()
                desc = input("Описание (Enter - пропустить): ").strip()
                
                # Дедлайн?
                has_deadline = input("Установить дедлайн? (y/n): ").lower()
                due_date = None
                if has_deadline == 'y':
                    days = int(input("Через сколько дней? "))
                    due_date = datetime.now() + timedelta(days=days)
                
                self.card_manager.create_card(selected_list['id'], title, desc, due_date)
        except ValueError:
            print("❌ Неверный ввод")
    
    def move_card_flow(self):
        """Перемещение карточки"""
        if not self.lists:
            print("❌ Нет списков!")
            return
        
        # Показываем карточки из всех списков
        all_cards = []
        print("\n📌 Все карточки:")
        for list_name, lst in self.lists.items():
            cards = self.card_manager.get_cards(lst['id'])
            for card in cards:
                all_cards.append(card)
                print(f"   [{list_name}] {card['name']} (ID: {card['id'][:8]}...)")
        
        if not all_cards:
            print("❌ Нет карточек для перемещения")
            return
        
        card_id = input("\nВведите ID карточки: ").strip()
        
        print("\nКуда переместить?")
        list_names = list(self.lists.keys())
        for i, name in enumerate(list_names, 1):
            print(f"   {i}. {name}")
        
        try:
            choice = int(input("Выберите список: "))
            if 1 <= choice <= len(list_names):
                target_list = self.lists[list_names[choice-1]]
                self.card_manager.move_card(card_id, target_list['id'])
        except ValueError:
            print("❌ Неверный ввод")
    
    def set_due_date_flow(self):
        """Установка дедлайна"""
        # Показываем карточки без дедлайна
        print("\n📌 Карточки без дедлайна:")
        cards_without_due = []
        
        for list_name, lst in self.lists.items():
            cards = self.card_manager.get_cards(lst['id'])
            for card in cards:
                if not card.get('due'):
                    cards_without_due.append(card)
                    print(f"   [{list_name}] {card['name']} (ID: {card['id'][:8]}...)")
        
        if not cards_without_due:
            print("❌ Нет карточек без дедлайна")
            return
        
        card_id = input("\nВведите ID карточки: ").strip()
        days = int(input("Через сколько дней? "))
        due_date = datetime.now() + timedelta(days=days)
        
        self.card_manager.set_due_date(card_id, due_date)
    
    def show_cards_flow(self):
        """Показать все карточки"""
        if not self.lists:
            print("❌ Нет списков!")
            return
        
        print("\n📋 ВСЕ КАРТОЧКИ:")
        for list_name, lst in self.lists.items():
            print(f"\n📌 {list_name}:")
            cards = self.card_manager.get_cards(lst['id'])
            if not cards:
                print("   (пусто)")
    
    def board_workflow(self):
        """Работа с выбранной доской"""
        while self.current_board:
            self.print_board_menu()
            choice = input("Ваш выбор: ")
            
            if choice == '0':
                self.current_board = None
                break
            elif choice == '1':
                self.add_list_flow()
            elif choice == '2':
                self.list_manager.get_lists(self.current_board['id'])
            elif choice == '3':
                self.add_card_flow()
            elif choice == '4':
                self.move_card_flow()
            elif choice == '5':
                self.set_due_date_flow()
            elif choice == '6':
                self.show_cards_flow()
            else:
                print("❌ Неверный выбор")
            
            input("\nНажмите Enter для продолжения...")
    
    def run(self):
        """Запуск CLI приложения"""
        while True:
            self.print_menu()
            choice = input("Ваш выбор: ")
            
            if choice == '0':
                print("\n👋 До свидания!")
                break
            elif choice == '1':
                self.create_board_flow()
            elif choice == '2':
                self.select_board_flow()
            elif choice == '3':
                self.board_manager.get_boards()
            else:
                print("❌ Неверный выбор")
            
            input("\nНажмите Enter для продолжения...")

# ============= ЗАПУСК =============
if __name__ == "__main__":
    print("\n🎯 TRELLO TASK MANAGER")
    print("="*50)
    print("\nИнструкция по получению ключей:")
    print("1. API Key: https://trello.com/power-ups/admin")
    print("2. Token: https://trello.com/1/authorize?expiration=never&scope=read,write,account&response_type=token&key=ТВОЙ_API_КЛЮЧ")
    print()
    
    API_KEY = input("🔑 Введите API Key: ").strip()
    TOKEN = input("🔐 Введите Token: ").strip()
    
    try:
        app = TrelloCLI(API_KEY, TOKEN)
        app.run()
    except Exception as e:
        print(f"❌ Ошибка: {e}")
````

````python
# 📊 ШАГ 5: Расширенная версия с дополнительной статистикой
import requests
from collections import Counter

def get_top_authors_advanced():
    """Расширенная версия с дополнительной статистикой"""
    
    # Загружаем данные
    posts = requests.get('https://jsonplaceholder.typicode.com/posts').json()
    users = {u['id']: u['name'] for u in 
             requests.get('https://jsonplaceholder.typicode.com/users').json()}
    
    # Собираем статистику
    author_stats = {}
    for post in posts:
        user_id = post['userId']
        if user_id not in author_stats:
            author_stats[user_id] = {
                'name': users[user_id],
                'total_length': 0,
                'post_count': 0,
                'avg_length': 0,
                'posts': []
            }
        
        author_stats[user_id]['total_length'] += len(post['body'])
        author_stats[user_id]['post_count'] += 1
        author_stats[user_id]['posts'].append({
            'id': post['id'],
            'title': post['title'],
            'length': len(post['body'])
        })
    
    # Вычисляем среднюю длину
    for user_id in author_stats:
        stats = author_stats[user_id]
        stats['avg_length'] = stats['total_length'] / stats['post_count']
    
    # Сортируем по общей длине
    sorted_authors = sorted(
        author_stats.values(),
        key=lambda x: x['total_length'],
        reverse=True
    )
    
    return sorted_authors[:3]

# Запускаем расширенную версию
top_authors = get_top_authors_advanced()

print("🏆 ТОП-3 АВТОРА ПО ДЛИНЕ ПОСТОВ")
print("="*60)
for i, author in enumerate(top_authors, 1):
    print(f"\n{i}. {author['name']}")
    print(f"   📊 Всего символов: {author['total_length']:,}")
    print(f"   📝 Количество постов: {author['post_count']}")
    print(f"   📈 Средняя длина поста: {author['avg_length']:.0f}")
    print(f"   🔥 Самый длинный пост: {max(a['length'] for a in author['posts'])} символов")
````

````python
# 🎯 ШАГ 6: Визуализация результатов
import requests
from collections import Counter

def get_top_authors_with_visual():
    """Версия с визуализацией результатов"""
    
    # Загружаем данные
    posts = requests.get('https://jsonplaceholder.typicode.com/posts').json()
    users = {u['id']: u['name'] for u in 
             requests.get('https://jsonplaceholder.typicode.com/users').json()}
    
    # Подсчитываем
    author_lengths = Counter()
    for post in posts:
        author_lengths[post['userId']] += len(post['body'])
    
    # Получаем топ-3
    top_3 = author_lengths.most_common(3)
    result = [(users[user_id], length) for user_id, length in top_3]
    
    # Визуализируем
    print("\n" + "="*60)
    print("🏆 ТОП-3 АВТОРА ПО ДЛИНЕ ПОСТОВ")
    print("="*60)
    
    max_length = max(length for _, length in result)
    
    for i, (name, length) in enumerate(result, 1):
        # Создаем прогресс-бар
        bar_length = int(40 * length / max_length)
        bar = "█" * bar_length + "░" * (40 - bar_length)
        
        print(f"\n{i}. {name}")
        print(f"   {bar} {length:,} символов")
    
    # Дополнительная статистика
    total_posts = len(posts)
    total_chars = sum(author_lengths.values())
    avg_per_author = total_chars / len(users)
    
    print("\n" + "="*60)
    print("📊 ОБЩАЯ СТАТИСТИКА")
    print("="*60)
    print(f"📝 Всего постов: {total_posts}")
    print(f"📝 Всего символов: {total_chars:,}")
    print(f"👥 Всего авторов: {len(users)}")
    print(f"📈 Среднее на автора: {avg_per_author:.0f} символов")
    
    return result

# Запускаем
top_authors = get_top_authors_with_visual()
````

````python
# 💡 ШАГ 7: Решение с обработкой ошибок
import requests
from collections import Counter
from time import sleep

def get_top_authors_safe(max_retries=3):
    """Версия с обработкой ошибок и повторными попытками"""
    
    for attempt in range(max_retries):
        try:
            # Загружаем посты
            print(f"📡 Попытка {attempt + 1}: загрузка данных...")
            posts_response = requests.get(
                'https://jsonplaceholder.typicode.com/posts',
                timeout=5
            )
            posts_response.raise_for_status()
            posts = posts_response.json()
            
            # Загружаем пользователей
            users_response = requests.get(
                'https://jsonplaceholder.typicode.com/users',
                timeout=5
            )
            users_response.raise_for_status()
            users = users_response.json()
            
            # Создаем словарь пользователей
            user_names = {user['id']: user['name'] for user in users}
            
            # Подсчитываем длину постов
            author_lengths = Counter()
            for post in posts:
                author_lengths[post['userId']] += len(post['body'])
            
            # Получаем топ-3
            top_3 = author_lengths.most_common(3)
            result = [(user_names[user_id], length) for user_id, length in top_3]
            
            print("✅ Данные успешно загружены и обработаны")
            return result
            
        except requests.exceptions.RequestException as e:
            print(f"❌ Ошибка загрузки: {e}")
            if attempt < max_retries - 1:
                print(f"🔄 Повторная попытка через {attempt + 1} секунд...")
                sleep(attempt + 1)
            else:
                print("❌ Не удалось загрузить данные после всех попыток")
                return []
        except Exception as e:
            print(f"❌ Неожиданная ошибка: {e}")
            return []

# Запускаем безопасную версию
top_authors = get_top_authors_safe()

if top_authors:
    print("\n🏆 ТОП-3 АВТОРА ПО ДЛИНЕ ПОСТОВ:")
    for i, (name, length) in enumerate(top_authors, 1):
        print(f"{i}. {name}: {length:,} символов")
else:
    print("❌ Не удалось получить результаты")
````
